#### 1.0 Classes
##### Definição:
Classes são moldes (modelos, formas, gabaritos, plantas) que definem a estrutura (atributos) e os comportamentos (métodos) dos elementos do programa. Logo, as classes não ocupam espaço na memória, não possui dados reais, e os objetos possui valores específicos para os atributos definidos pela classe. 

#### 1.1 Objetos

##### Definição:
Objetos são instâncias concretas criadas a partir do molde (forma, planta, esquemático), passando a existir na memória com dados e funções próprios.

A analogia mais comumente usada no entendimento de **classe** e **objeto** é a planta de uma casa, **classe**, e a casa construída, **objeto**. Os construtores usam a planta para definir como a casa será, mas só se pode morar na casa construída.  

Na linguagem Python tudo que se usa em um programa é objeto, ou seja, de constantes a variáveis, de funções a classes, ... tudo é criado e gerenciado como um objeto na linguagem Python.

##### Exempo:
* Classe: `Carro`
  * Atributos: `modelo` (string), `cor` (string), ``ano`` (inteiro).
  * Métodos: `acelerar()`, `frear()`.  
  
Ao criar objetos a partir da classe **Carro**, geramos entidades únicas:
* Objeto 1: `meuCarro`
  * Atributos: `modelo` = "Fusca", `cor` = "Azul", `ano` = 1970.
* Objeto 2: `carroDoAmigo`
  * Atributos: `modelo` = "Ferrari", `cor` = "Vermelha", `ano` = 2026.

Ambos têm a mesma origem e compartilham os métodos da classe **Carro**, mas seus estados (atributos, dados) são independentes, armazenados em locais distintos de memória.

#### 1.2 Métodos


Em Python, as classes são criadas com a palavra-chave `class`, e os objetos são gerados chamando essa classe como se fosse uma função, e eventualmente, passando valores iniciais para os atributos do objeto.

### <span style="color: #0040ff;" id="Exemplo">2. Exemplo Prático

```python
# Definindo a Classe
class Carro:
    # O método __init__ é o construtor do objeto, e define os atributos que cada objeto terá
    def __init__(self, modelo, cor, ano):
        self.modelo = modelo      # atributo
        self.cor = cor            # atributo
        self.ano = ano            # atributo

    # Definindo um método (comportamento)
    def acelerar(self):
        print(f"O modelo {self.modelo} está acelerando, Vrummm!")

# Criando os Objetos (Instanciação)
meuCarro = Carro("Fusca", "Azul", 1970)
carroDoAmigo = Carro("Ferrari", "Vermelha", 2026)

# Acessando dados e executando ações
print(meuCarro.cor)              # Saída: Azul
carroDoAmigo.acelerar()          # Saída: O Ferrari está acelerando: Vrummm!
```

##### Detalhes importantes do Python:  
* Método **\_\_init\_\_**: Este é o método construtor da classe, aquele que cria o objeto (instância a classe) na memória. Ele roda automaticamente sempre que se cria um objeto novo para inicializar os atributos.
* O Parâmetro ***self***: Representa o objeto específico que está executando aquela ação. Ele conecta as variáveis e métodos à instância atual.

### <span style="color: #0040ff;" id="Heranca">3. Herança

Herança é o mecanismo da POO que permite que uma nova classe (classe filha) reutilize e estenda os atributos e métodos de uma classe já existente (classe mãe). Isso evita a repetição de código e ajuda a criar uma estrutura hierárquica bem clara.

Seguindo o exemplo anterior, classe `Carro` podemos criar uma classe filha chamada `CarroEletrico` que herda tudo da classe Carro, mas ganha características (atributos) próprias, como o tamanho da bateria.

```python
# Classe Filha (Subclasse), herdando da classe Carro
class CarroEletrico(Carro):
    def __init__(self, modelo, cor, ano, cap_bateria):
        # a chamada super().__init__ reaproveita o construtor da classe mãe
        super().__init__(modelo, cor, ano)
        self.cap_bateria = cap_bateria  # atributo exclusivo da classe filha

    # Método exclusivo da classe filha
    def carregar_bateria(self):
        print(f"O modelo {self.modelo} está carregando a bateria de {self.cap_bateria} kWh.")

# Criando um objeto da classe filha
meuEletrico = CarroEletrico("BYD Song Plus", "Branco", 2025, 57)

# O objeto pode usar os métodos da classe mãe...
meuEletrico.acelerar()          # Saída: O modelo BYD Song Plus está acelerando, Vrummm!

# ...e também os seus próprios métodos
meuEletrico.carregar_bateria()  # Saída: O modelo BYD Song Plus está carregando a bateria de 57 kWh.
```

```python
class Pai:
    def habilidade_marcenaria(self):
        print("Sabe fazer móveis de madeira.")

class Mae:
    def habilidade_culinaria(self):
        print("Sabe cozinhar muito bem.")


class Filho(Pai, Mae):              # Herança múltipla: Filho herda de Pai e Mae
    pass

# Testando
eu = Filho()
eu.habilidade_marcenaria()          # Herdado do pai
eu.habilidade_culinaria()           # Herdado da mãe
```

##### Conceitos-Chave da Herança em Python:
* Sintaxe de Parênteses: Para herdar os atributos e métodos de uma classe mãe, é preciso indicar a classe mãe entre parênteses depois do nome da classe filha na declaração da classe: `class ClasseFilha(ClasseMae):`.
* A Função `super()`: Serve para chamar métodos da classe mãe. É usada principalmente no construtor `__init__()` para não precisar reescrever a lógica de criação dos atributos básicos (modelo, cor, ano).

### <span style="color: #0040ff;" id="Polimorfismo">4. Polimorfismo

**Polimorfismo** significa "muitas formas". Na POO, é o conceito que permite que classes filhas usem um mesmo método herdado da classe mãe, mas adaptem o comportamento para a sua própria realidade.  

No exemplo anterior, o método `acelerar()` de um carro comum faz o som "Vrummm!". Mas um carro elétrico é silencioso. Com o polimorfismo, a classe filha pode sobrescrever (fazer um *override*) desse método para mudar seu funcionamento.

```python
# Classe Filha 1 (Polimorfismo por Sobrescrita - override)
class CarroEletrico(Carro):
    # Sobrescrevendo o método acelerar da classe mãe
    def acelerar(self):
        print(f"O modelo {self.modelo} está acelerando silenciosamente, Zuummm!")

# Classe Filha 2 (Outra variação)
class CarroCorrida(Carro):
    # Sobrescrevendo o método acelerar de outra forma
    def acelerar(self):
        print(f"O modelo {self.modelo} de corrida está decolando, VRRRAAAUUU!")

# Criando os objetos
fusca = Carro("Fusca", "Azul", 1970)
tesla = CarroEletrico("Tesla Model 3", "Branco", 2026)
formula1 = CarroCorrida("Ferrari-F1", "Vermelho", 2026)

# Chamando o MESMO método em objetos diferentes
fusca.acelerar()     # Saída: O modelo Fusca está acelerando: Vrummm!
tesla.acelerar()     # Saída: O modelo Tesla Model 3 está acelerando silenciosamente, Zuummm!
formula1.acelerar()  # Saída: O modelo Ferrari-F1 de corrida está decolando, VRRRAAAUUU!
```

O **polimorfismo** permite tratar objetos diferentes de forma unificada. Veja como podemos colocar todos os carros em uma lista e acelerar todos de uma vez, sem que o código precise saber o tipo exato de cada carro:

```python
lista_de_carros = [fusca, tesla, formula1]

for carro in lista_de_carros:
    carro.acelerar()  # Cada objeto responde do seu próprio jeito!
```

### <span style="color: #0040ff;" id="Encapsulamento">5. Encapsulamento

**Encapsulamento** é o pilar da POO que serve para proteger e esconder os dados internos de um objeto. Ele impede que um código de fora da classe altere atributos sensíveis diretamente, garantindo que o objeto controle suas próprias regras de negócio. Em linguagens como Java ou C#, existem palavras-chave como `private` e `public` para se fazer esse controle. O Python não tem esses bloqueios rígidos no sistema, mas usa uma convenção de nomenclatura baseada em sublinhados (um sublinhado, '_', ou dois sublinhados seguidos, '__') para indicar o nível de acesso aos atributos: **protegido** ou **privado**.  

##### Níveis de Acesso em Python
* **Público** (sem sublinhado): Pode ser acessado de qualquer lugar.  
* **Protegido** (um sublinhado _): Indica que o atributo não deve ser acessado fora da classe ou de suas classes filhas (um aviso amigável para outros desenvolvedores).
* **Privado** (dois sublinhados __): O Python tranca o acesso direto de fora da classe usando um mecanismo chamado *Name Mangling* (transformação de nome).

##### Exemplo:
Imagine um `CarroEletrico` onde a carga da bateria não pode ser alterada diretamente para qualquer valor (como um valor negativo ou acima de 100%). Usamos métodos *Getters* e *Setters* para ler e modificar esse dado com segurança.

```python
class CarroEletrico:
    def __init__(self, modelo, cap_bateria):
        self.modelo = modelo
        self.__carga_bateria = 50    # atributo privado usando __ (Modo seguro)

    # GETTER: Método para permitir ler o valor com segurança
    def exibir_carga(self):
        return f"A bateria do {self.modelo} está em {self.__carga_bateria}%."

    # SETTER: Método para alterar o valor aplicando regras de validação
    def recarregar(self, quantidade):
        if quantidade < 0:
            print("Erro: Não é possível recarregar valores negativos!")
        elif self.__carga_bateria + quantidade >= 100:
            self.__carga_bateria = 100
            print("Bateria totalmente carregada (100%)!")
        else:
            self.__carga_bateria += quantidade
            print(f"Recarregado! Carga atual: {self.__carga_bateria}%")

# Testando o Encapsulamento
meu_tesla = CarroEletrico("Tesla", 100)

# Forma correta de alterar o dado com validação (via Setter)
meu_tesla.recarregar(-20)        # Saída: Erro: Não é possível recarregar valores negativos!
meu_tesla.recarregar(10)
meu_tesla.recarregar(60)

# Forma correta de ler o dado (via Getter)
print(meu_tesla.exibir_carga())  # Saída: A bateria do Tesla está em 100%

# Tentar acessar diretamente gera um erro (AttributeError)
# print(meu_tesla.__carga_bateria) # Descomentar esta linha quebrará o código!
```

##### O jeito "Pythonico": @property  

Em Python do dia a dia, em vez de criar métodos como `get_carga()` ou `set_carga()`, usamos um recurso elegante chamado ***Decorators*** (**@property**). Ele faz com que um método se comporte visualmente como se fosse um atributo comum:

```python
class Smartphone:
    def __init__(self):
        self.__bateria = 100

    @property
    def bateria(self):              # método GETTER (lê o valor do atributo privado __bateria)
        return self.__bateria

    @bateria.setter
    def bateria(self, valor):       # método SETTER (estipula o valor para o atributo privado __bateria)
        if 0 <= valor <= 100:
            self.__bateria = valor

celular = Smartphone()
celular.bateria = 85    # Usa o 'setter' por baixo dos panos de forma limpa!
print(celular.bateria)  # Saída: 85
```

Agora você já conhece a base teórica e prática de POO em Python: Classes, Objetos, Herança, Polimorfismo e Encapsulamento.  
Vamos praticar esse conhecimento com um pequeno desafio guiado, criando um sistema simples de conta bancária.

```python
class ContaBancaria:
    def __init__(self, titular, numero_conta, saldo_inicial=0.0):
        self.titular = titular          # atributo público
        self._numero = numero_conta     # atributo protegido
        self.__saldo = saldo_inicial    # atributo privado (protegido contra alterações diretas)

    # GETTER para o saldo (permite ver o saldo, mas não alterar diretamente)
    @property
    def saldo(self):
        return self.__saldo

    # Método para depositar dinheiro (com validação)
    def depositar(self, valor):
        if valor > 0:
            self.__saldo += valor
            print(f"Depósito de R$ {valor:.2f} realizado com sucesso!")
        else:
            print("Operação inválida: O valor do depósito deve ser maior que zero.")

    # Método para sacar dinheiro (com validação de saldo)
    def sacar(self, valor):
        if valor <= 0:
            print("Operação inválida: O valor do saque deve ser maior que zero.")
        elif valor > self.__saldo:
            print(f"Saldo insuficiente! Seu saldo atual é de R$ {self.__saldo:.2f}")
        else:
            self.__saldo -= valor
            print(f"Saque de R$ {valor:.2f} realizado com sucesso!")

    # Método para exibir o extrato atualizado
    def exibir_extrato(self):
        print(f"\n--- Extrato ---")
        print(f"Conta: {self._numero} | Titular: {self.titular}")
        print(f"Saldo Atual: R$ {self.__saldo:.2f}")
        print("-" * 15)

# --- Usando o Sistema Bancário ---

# Criando a conta do cliente José, com saldo inicial de R$ 500
conta_jose = ContaBancaria("José", "12345-6", 500.0)

# Tentando visualizar e alterar o saldo diretamente (o encapsulamento protege aqui)
print(conta_jose.saldo)       # Saída: 500.0 (acesso permitido via getter)
conta_jose.__saldo = 1000000  # Erro! Isso não altera o saldo real da conta.

# Realizando operações legítimas
conta_jose.depositar(200.0)   # Sucesso!
conta_jose.sacar(100.0)       # Sucesso!

# Tentando quebrar as regras do negócio
conta_jose.sacar(800.0)       # Bloqueado: Saldo insuficiente
conta_jose.depositar(-50)     # Bloqueado: Valor inválido

# Exibindo o resultado final
conta_jose.exibir_extrato()
```

##### Exercício:  
A partir do código da célula anterior, crie uma classe `ContaCorrente` (que possui um limite de cheque especial), usando Herança e Polimorfismo para adequar as mensagens do sistema bancário.

```python
# Gabarito

# Reutilizando a Classe Mãe (Superclasse)
class ContaBancaria:
    def __init__(self, titular, numero_conta, saldo_inicial=0.0):
        self.titular = titular
        self._numero = numero_conta
        self._saldo = saldo_inicial  # Mudamos para um sublinhado para que as classes filhas possam acessar diretamente

    @property
    def saldo(self):
        return self._saldo

    def depositar(self, valor):
        if valor > 0:
            self._saldo += valor
            print(f"Depósito de R$ {valor:.2f} realizado com sucesso!")
        else:
            print("Operação inválida.")

    def sacar(self, valor):
        if 0 < valor <= self._saldo:
            self._saldo -= valor
            print(f"Saque de R$ {valor:.2f} realizado!")
        else:
            print("Saldo insuficiente ou valor inválido.")

# Classe Filha (Subclasse)
class ContaCorrente(ContaBancaria):
    def __init__(self, titular, numero_conta, saldo_inicial=0.0, limite=500.0):
        # super() aciona o construtor da classe mãe para iniciar os atributos: titular, número e saldo
        super().__init__(titular, numero_conta, saldo_inicial)
        self.limite = limite      # atributo exclusivo da Conta Corrente

    # POLIMORFISMO: Sobrescrita (override) do método 'sacar' para incluir a lógica do limite do cheque especial
    def sacar(self, valor):
        if valor <= 0:
            print("Operação inválida: O valor do saque deve ser maior que zero.")
        # Verifica se o valor está dentro do saldo disponível + o limite do cheque especial
        elif valor > (self._saldo + self.limite):
            print(f"Saldo e limite insuficientes! Limite total disponível: R$ {self._saldo + self.limite:.2f}")
        else:
            self._saldo -= valor
            print(f"Saque de R$ {valor:.2f} realizado com sucesso (Uso do limite permitido)!")

    # Modificando o extrato para exibir também o limite disponível
    def exibir_extrato(self):
        print(f"\n--- Extrato Conta Corrente ---")
        print(f"Conta: {self._numero} | Titular: {self.titular}")
        print(f"Saldo em Conta.............: R$ {self._saldo:7.2f}")
        print(f"Limite de Cheque Especial:.: R$ {self.limite:7.2f}")
        print(f"Total Disponível para Saque: R$ {self._saldo + self.limite:7.2f}")
        print("-" * 30)

# --- Usando a classe 'Conta Corrente' ---

# Criando uma conta corrente com R$ 100 de saldo e R$ 500 de limite
cc_maria = ContaCorrente("Maria", "54321-0", 100.0, 500.0)

# Sacando um valor maior que o saldo, mas dentro do limite (R$ 100 saldo + R$ 300 limite)
cc_maria.sacar(400.0)  # SMariao! O saldo vai ficar em R$ -300.00

# Tentando sacar além do limite permitido
cc_maria.sacar(300.0)  # BMariaado! Restavam apenas R$ 200 de limite disponível

# Verificando o extrato atualizado
cc_maria.exibir_extrato()
```
