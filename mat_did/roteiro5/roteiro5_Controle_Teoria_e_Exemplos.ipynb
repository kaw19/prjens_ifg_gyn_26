## Widgets(componentes)

Quando se está ploando gráficos em python, é as vezes interessante que se possa alterar parâmetros na equação para se observar como ela afeta a saída do programa.

Porém não é conveniente que seja necessário voltar ao código, alterar os parâmetros, executar o código e esperar que se obtenha uma resposta, principalmente em códigos em que o tempo de resposta pode chegar aos segundos. Por isso o pacote ```ipywidgets``` pode ser muito interessante 

O pacote ```ipywidgets ``` é uma biblioteca Python desenvolvida para criar interfaces gráficas interativas diretamente em ambientes computacionais baseados em notebooks, como o Jupyter Notebook e o JupyterLab

Seu principal objetivo é permitir que o usuário interaja com programas e visualizações de maneira dinâmica, sem a necessidade de desenvolver interfaces complexas utilizando ferramentas externas.

A biblioteca fornece diversos componentes gráficos, chamados de widgets, que possibilitam modificar parâmetros de programas enquanto eles estão em execução, tornando a análise de dados, simulações matemáticas e experimentos computacionais muito mais intuitivos.

### 1.1 Widget-sliders (Componentes deslizantes)

Os sliders(deslizantes) são uma das formas mais intuitivas de se modificar parâmetros apartir de interfaes gráficas, pois permitem que se altere valores de forma bem controlada e dentro de uma faixa definida com intervalos também alteráveis.

Aqui está um exemplo:

````python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

# Cria os dados do eixo X
x = np.linspace(0, 10, 1000)

# Função que atualiza o gráfico
def atualizar(amplitude, frequencia, fase):        #atualizar é a função que permite que os valores sejam constantemente modificados.
    y = amplitude * np.sin(frequencia * x + fase)

    plt.figure(figsize=(8,4))                   #Gráfico de uma senóide
    plt.plot(x, y)
    plt.title("Senoide com ajuste em tempo real")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(True)
    plt.ylim(-10, 10)
    plt.show()

# Criação dos sliders
slider_amplitude = widgets.FloatSlider(         #Criando um slider para valores reais
    value=1,                                    #Valor inicial
    min=0,                                      #Valor mínimo
    max=10,                                     #Valor máximo
    step=0.1,                                   #Tamanho dos intervalos
    description='Amplitude'                     #Nome quse dará ao slider
)

slider_frequencia = widgets.FloatSlider(
    value=1,
    min=0.1,
    max=10,
    step=0.1,
    description='Freq.'
)

slider_fase = widgets.FloatSlider(
    value=0,
    min=0,
    max=2*np.pi,
    step=0.1,
    description='Fase'
)

# Liga os sliders à função
widgets.interactive(                    #Função que relaciona as variaeis do valor do slider 
    atualizar,                          #com as variáveis da equação
    amplitude=slider_amplitude,
    frequencia=slider_frequencia,
    fase=slider_fase
)
````

Mas o slider não é limitado a horizontal, se pode fazer sliders verticais também mudando sua orientação. 

````python
widgets.FloatSlider(value=5, min=0, max=10, orientation='vertical')
````
