### 1. Definição de Sistema de Controle

É qualquer configuração que gerencia o comportamento de uma máquina ou processo.  
Todo **Sistema de Controle** usa uma entrada como referência, afeta o comportamento de um sistema e tenta produzir uma saída desejada (*setpoint*).

### 2. Exemplos de Sistema de Controle
**Termostato** de uma torradeira, temporizador do forno microondas, controle de cruzeiro gerenciando a velocidade do carro, ...  

### 3. Tipos de Sistema de Controle

#### 3.1 Sistema em Malha Aberta (*Open-loop*)
* __Definição 1__: É todo sistema no qual a ação de controle é independente da saída, ou seja, a saída não influencia a ação de controle.  

* __Definição 2__: É uma estrutura de controle na qual a entrada é independente da saída, operando sem *feedback* para corrigir o comportamento. Baseia-se apenas em entradas calibradas e temporização predefinida, tornando-o simples e económico, mas vulnerável a perturbações e incapaz de autocorreção.  

O __controle em malha aberta__ é adequado apenas quando a dinâmica do sistema é altamente previsível e os distúrbios externos são pequenos.

A entrada executa o processo (temporizador da torradeira), e não existe nenhuma retroalimentação (*feedback*).  

No forno microondas o usuário configura o tempo de acionamento e a taxa de potência de transferência de calor, nenhuma retroalimentação é usada nesse processo de cocção, ou seja, se o alimento esteja congelado e o tempo estipulado não for longo o suficiente, ainda assim o forno é desligado no prazo estipulado.

#### 3.2 Sistema em Malha Fechada
* Definição 1: É todo sistema  no qual a ação de controle depende da saída, ou seja, a saída causa um efeito direto na ação de controle do sistema.

Trata-se de sistemas mais acurado e auto-corregível, com comparação contínua da saída em relação ao *setpoint*, para ajuste da variável controlada.

**Realimentação**: monitora a saída e ajusta a entrada para alcançar/permanecer no alvo estabelecido pelo *setpoint*.

**Exemplo**: Termostato que liga/desliga o aquecedor/resfriador baseado na temperatura do ambiente e a temperatura programada (*setpoint*).
A partir do erro o **controlador** toma decisão sobre o que fazer para alcançar o *setpoint*, aumentando ou diminuindo a intensidade da variável controlada. Existem vários tipos de controladores, mas o mais comum é o **controlador PID**.

### <span style="color: #840371;">5. Motor DC (Revisão)</span>

O motor de corrente contínua é constituído de **enrolamento de armadura**, **enrolamento de campo** ou **imã permanente**, **comutador** e  **escovas**.

#### <span style="color: #840371;">5.1 Funcionamento</span>
A tensão aplicada à armadura ($V_a$) força a corrente de armadura ($i_a$) a circular pelos condutores do rotor. O comutador inverte o sentido da corrente nas bobinas à medida que elas giram, garantindo que o torque seja unidirecional.

A resistência e a indutância da armadura afetam a dinâmica do motor. Escovas de grafite ou material condutor são usadas para transferir essa corrente para o rotor. A comutação de alta frequência no controle de tensão média da armadura permite um controle preciso da velocidade.

<div align="center"><img src="rot3_fig7a.png" alt="Desenho do Estator e da Armadura de um Motor DC" width="300"> <img src="rot3_fig7b.png" alt="Desenho do Estator de um Motor DC" width="370"></div>

Os motores de maiores potência usam bobinas no **estator** para gerar campos magnéticos mais intensos que os providos por imãs permanentes. O **estator** é a parte fixa e externa de um motor elétrico, composta por um núcleo de chapas de aço silício (diminui perdas por calor, gerado por correntes parasitas, ou de Foucault: correntes induzidas em materiais condutores quando expostos a um campo magnético variável, circulando em *loops* fechados) e bobinas (enrolamentos). O estator de um motor DC é a parte fixa e externa que gera o campo magnético, essencial para a conversão de energia elétrica em mecânica. Composto por ímãs permanentes ou bobinas de campo (eletroímãs), ele envolve o rotor (induzido), interagindo para criar o torque necessário ao giro. 

Nos motores de indução (corrente alternada - AC), a principal função do estator é criar um campo magnético girante ao ser percorrido por corrente elétrica, interagindo com o rotor (parte móvel do motor) para gerar movimento giratório.

<div align="center"><img src="rotorbobinado.gif" alt="motor de indução com rotor bobinado" width="500"></div>

####  <span style="color: #840371;">5.2 Características do Estator</span>
**Parte Estacionária**: Diferente do rotor, ele permanece estático, fixado à carcaça do motor.  
**Campo Magnético**: Quando energizado, o estator cria o campo magnético que induz o giro no rotor, convertendo energia elétrica em energia mecânica.  
**Componentes**: Constituído por um núcleo magnético de ferro laminado (para reduzir perdas por correntes parasitas) e enrolamentos de cobre.  
**Aplicações**: Essencial em motores de indução (assíncronos), motores síncronos e geradores (como alternadores de motocicletas que geram/convertem energia para o sistema elétrico).  

Em motores de alta eficiência, o **estator** é projetado para maximizar a interação magnética e minimizar a perda de energia.

#### <span style="color: #840371;">5.4 Equação Básica</span>
A corrente de Armadura é determinada por: $~~~I_a = (V_a - e_b)/R_a$  
onde:  
    $~~~~V_a$ - é a tensão da armadura,  
    $~~~~e_b$ - é a força contraeletromotriz (FCEM),  
    $~~~~R_a$ - é a resistência da armadura.  
    
#### <span style="color: #840371;">5.5 Controle da Velocidade Angular</span> ($\omega$)
Variar a tensão na armadura ($V_a$) é a principal forma de controlar a velocidade angular do motor, frequentemente realizada via modulação por largura de pulso (PWM - *Pulse Width Modulation*).

#### <span style="color: #840371;">5.6 Torque</span> ($T$)
O torque desenvolvido é diretamente proporcional à intensidade da corrente de armadura ($T \propto i_a$), o que significa que carga maior exige maior corrente.

#### <span style="color: #840371;">5.7 Características</span>
A corrente de armadura ($I_a$) pode ser muito alta na partida, pois a FCEM é zero ($~~~~e_b = 0$), exigindo limitadores de corrente em motores de grande porte.

### <span style="color: #0011fa;">7. Modelagem Matemática</span>
A modelagem de um motor de corrente contínua com excitação independente baseia-se na interação entre os domínios elétrico e mecânico.
Para simular o comportamento do motor (por exemplo, para controle PID), podemos modelar a relação entre a **tensão de entrada** (tensão de armadura, $V_a$) e a velocidade angular ($\omega$). Um modelo linear padrão assume mudanças na velocidade do motor com base em sinais de controle e constantes físicas como inércia ($J$) e atrito. 

#### <span style="color: #0011fa;">7.1 Principais Parâmetros</span>

#### <span style="color: #0011fa;">7.2 Equação Elétrica (Circuito da Armadura)</span>  
Aplicando a Lei das Tensões de Kirchhoff ao circuito da armadura:  

$$ V_a(t) = R_a . i_a(t) + L_a .\frac{di_a(t)}{dt} + e_a(t) $$

onde: 
    $e_a(t)$ é a força contra-eletromotriz (FCEM), que é proporcional à velocidade angular da armadura, $\omega(t)$:   
    $$e_a(t)=K_\phi.\omega(t)$$

#### <span style="color: #0011fa;">7.3 Equação Mecânica</span>
Baseada na Segunda Lei de Newton para Rotação, o torque eletromagnético ($T_{em}$) desenvolvido no eixo do motor, deve vencer a inércia do rotor, o atrito ($B$ é o coeficiente de atrito viscoso) e o torque da carga ($T_L$):  
$$T_{em}(t) = J \frac{d\omega(t)}{dt} + B .\omega(t) + T_L(t)$$

O torque eletromagnético é proporcional à corrente de armadura: $~~~~T_{em}(t) = K_\phi . i_a(t)$

### <span style="color: #00d4cd;">8. Simulação Computacional</span>

A simulação de um motor DC (CC - corrente contínua) pelo método de Euler consiste em discretizar as equações diferenciais (EDO's) que regem o comportamento elétrico e mecânico do motor, permitindo calcular a **velocidade angular** e a **corrente de armadura** em pequenos passos de tempo $h$.

````python
# Simulação da partida direta ou partida degrau (tensão constante aplicada repentinamente) de um motor DC
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros do Motor DC
Ra = 2.0             # Resistência da armadura (Ohm)
La = 0.05            # Indutância da armadura (Henry)
Kv = 0.1             # Constante de força contra-eletromotriz (V/(rad/s))
Kt = 0.1             # Constante de torque (N.m/A)
J = 0.01             # Inércia do rotor (kg.m^2)
B = 0.005            # Coeficiente de atrito viscoso (N.m/(rad/s))

# Parâmetros da Simulação
t_final = 10.0       # tempo final (s)
dt = 0.001           # infinitésimo de tempo (precisão da simulação)
t = np.arange(0, t_final, dt)
n = len(t)           # quantidade de pontos no eixo horizontal (abscissas)

# Inicialização de vetores
va = np.zeros(n)     # Tensão de armadura
va[100:] = 50        # Aplica 50 V após 0.1 s
ia = np.zeros(n)     # Corrente de armadura
w = np.zeros(n)      # Velocidade angular da armadura
torque = np.zeros(n) # Torque eletromagnético

# Simulação usando o Método de Euler para o cálculo da EDO
for k in range(1, n):
    e = Kv * w[k-1]  # Força Contra-Eletromotriz (FCEM)
    d_ia = (va[k-1] - Ra*ia[k-1] - e) / La # derivada da cor. de armadura em relação ao tempo
    ia[k] = ia[k-1] + d_ia * dt            # corrente de armadura anterior mais a variação (dia/dt)
    
    torque[k] = Kt * ia[k]                 # torque atual, no instante k
    d_w = (torque[k] - B * w[k-1]) / J     # variação da velocidae angular em um infinitésimo de tempo (dt)
    w[k] = w[k-1] + d_w * dt               # velocidade angular atual = veloc. anterior mais a variação (dw/dt)

# Visualização dos Resultados
plt.figure(figsize=(7, 5))
plt.subplot(2, 1, 1); plt.plot(t, ia)
plt.title('Corrente na Partida do Motor DC'); plt.ylabel('Corrente (A)'); plt.grid(True)

plt.subplot(2, 1, 2); plt.plot(t, w)
plt.title('Velocidade na Partida do Motor DC'); plt.xlabel('Tempo (s)'); plt.ylabel('Velocidade (rad/s)')  
plt.grid(True); plt.tight_layout()
plt.show()
````

#### <span style="color: #00d4cd;">8.1 Análise dos Gráficos</span>
Observe o pico de corrente inicial ($I_{start} ≈ V_a / R_a$). Compare o tempo que a velocidade angular leva para estabilizar com a constante de tempo mecânica. 

<span style="color: #06d1fa;"><b>Gráfico B</b></span>: Corrente de Armadura ($i_a$)  
- Pico de Partida: Em $t=0$, a corrente salta para 50 A ($I=V_a / R_a = 100/2$). Como a velocidade é zero, a FCEM também é zero, e apenas a resistência limita a corrente.  
- Regime Permanente: A corrente cai drasticamente à medida que o motor acelera. Sem carga, ela termina próxima de zero (apenas para vencer o atrito $B$). Com carga, ela estabiliza em um valor mais alto para manter o torque necessário.


#### <span style="color: #00d4cd;">8.2 Variação de Parâmetros</span>
1. Aumente o valor da inércia para $J = 0.1~kg\;m^2$. O que acontece com o tempo de aceleração?

````python
# Simulação da partida direta de um motor DC com maior constante inercial, J

# Parâmetros do Motor DC
Ra = 2.0             # Resistência da armadura (Ohm)
La = 0.05            # Indutância da armadura (Henry)
Kv = 0.1             # Constante de força contra-eletromotriz (V/(rad/s))
Kt = 0.1             # Constante de torque (N.m/A)
J = 0.1              # <-- Inércia do rotor (kg.m^2)
B = 0.005            # Coeficiente de atrito viscoso (N.m/(rad/s))

# Parâmetros da Simulação
t_final = 10.0       # tempo final (s)
dt = 0.001           # infinitésimo de tempo (precisão da simulação)
t = np.arange(0, t_final, dt)
n = len(t)           # quantidade de pontos no eixo horizontal (abscissas)

# Inicialização de vetores
va = np.zeros(n)     # Tensão de armadura
va[100:] = 100       # Aplica 100 V após 0.1 s
ia = np.zeros(n)     # Corrente de armadura
w = np.zeros(n)      # Velocidade angular
torque = np.zeros(n)

# Simulação (Método de Euler)
for k in range(1, n):
    e = Kv * w[k-1]                        # Força Contra-Eletromotriz (FCEM)
    d_ia = (va[k-1] - Ra*ia[k-1] - e) / La # derivada da cor. de armadura em relação ao tempo
    ia[k] = ia[k-1] + d_ia * dt            # cor. de armadura anterior mais a variação (dia/dt)
    
    torque[k] = Kt * ia[k]                 # torque atual, no instante k
    d_w = (torque[k] - B * w[k-1]) / J     # variação da velocidae angular em um infinitésimo de tempo (dt)
    w[k] = w[k-1] + d_w * dt               # velocidade angular atual = veloc. anterior mais a variação (dw/dt)

# Visualização dos Resultados
plt.figure(figsize=(7, 5))
# noqa: E702 (comentário usado no final de uma linha de código Python para instruir linters -- ferramenta de análise estática de código 
#             que verifica o código-fonte em busca de erros de sintaxe, bugs, inconsistências de estilo e violações de boas práticas 
#             de programação -- a ignorar o erro E702: "multiple statements on one line (semicolon))
plt.subplot(2, 1, 1); plt.plot(t, ia)      # noqa: E702
plt.title('Corrente na Partida do Motor DC'); plt.ylabel('Corrente (A)'); plt.grid(True)  # noqa: E702

plt.subplot(2, 1, 2); plt.plot(t, w)       # noqa: E702
plt.title('Velocidade na Partida do Motor DC'); plt.xlabel('Tempo (s)'); plt.ylabel('Velocidade (rad/s)')  # noqa: E702
plt.grid(True); plt.tight_layout()         # noqa: E702
plt.show()
````

1. O que acontece ao aumentar a Inércia ($J$)?  
**Resposta**: O sistema torna-se "mais lento" para responder. O tempo de subida da velocidade aumenta significativamente, mas o valor final da velocidade de regime e o pico de corrente permanecem os mesmos. A inércia afeta apenas a dinâmica transitória, não o equilíbrio estático.

1. O que acontece ao reduzir a Resistência ($R_a$)?  
**Resposta**: O pico de corrente de partida aumenta perigosamente. Se $R_a$ cair para 0.5 Ω, o pico saltaria de 50 A para 200 A. Isso demonstra por que motores de grande porte exigem reostatos de partida ou *soft-starters* para não danificar os enrolamentos.

#### <span style="color: #00d4cd;">8.3 Conclusão</span>
o motor CC é um sistema acoplado onde a velocidade mecânica ($\omega$) atua como uma realimentação (*feedback*) negativa na malha elétrica através da FCEM (Força Contra-EletroMotriz), limitando a corrente de armadura ($I_a$) em regime permanente.

Altere o _script_ da célula de código aplicada na simulação do motor DC para usar a seguinte função:
``` python
def metodo_euler(f, t0, y0, h, tn):
    """ Calcula dy/dt pelo método de Euler
    f : Função que define a EDO
    t0: Valor do tempo inicial
    y0: Condições iniciais
    h : Tamanho do passo
    tn: Valor do tempo final
    """
    t = np.arange(t0, tn + h, h)
    y = np.zeros(len(t))
    y[0] = y0
    for i in range(0, len(t) - 1):
        # regra de atualização: y_prox = y_corr + h * f(t_corr, y_corr)
        y[i+1] = y[i] + h * f(t[i], y[i])
        
    return t, y


### Anexo: Método de Euler
Trata-se de uma técnica numérica que aproxima a derivada $\frac{dx}{dt}$ por $\frac{x(k+1)-x(k)}{h}$.  

* Equação diferencial elétrica: $$V_a(t) = R_a.i(t) + L_a.\frac{di(t)}{dt} + e_a(t)$$  
Sabemos que a FCEM é proporcional à velocidade angular: $e_a(t) = K_\phi . \omega(t)$  
Logo: $$\frac{di(t)}{dt} = [V_a(t) - R_a.i(t) - K_\phi . \omega(t)]/L_a$$  

Aplicando o Método de Euler à equação diferencial elétrica, temos:
$$\boxed{\color{blue}i_a(k+1) = i(k) + h.[V_a(k) - R_a.i_a(k) - K_.\omega(k)]}$$

* Equação diferencial mecânica: $$T_{em}(t) = J.\frac{d\omega(t)}{dt} + B.\omega{t} + T_L(t)$$
Sabemos que o Torque eletromagnético é proporcional à corrente elétrica: $T_{em}(t) = K_\phi . i(t)$  
Logo: $$\frac{d\omega(t)}{dt} = [K_\phi .i(t) - B. \omega(t) - T_L(t)]/J$$ 

Aplicando o Método de Euler à equação diferencial mecânica, temos:  
$$\boxed{\color{blue}\omega(k+1) = \omega(k) + h.[K_\phi .i(k) + B.\omega(k) + T_L(k)]}$$

#### Exemplo
EDO: $i'_a(0) = 1$ e passo $h = 0.1$.  

Parâmetros do Motor: $K_\phi = 0.1,\;\; L_a = 0.05 H $  

Iteração 1:  
* $ k = 0, i_a(0) = 1$
* $ i_a(1) = i_a(0) + h . [V_a(0) - R_a.i_a(0) - K_\phi.\omega(0)]/L_a$
* $ \omega(1) = \omega(0) + h.[K.i_a(0) + B.\omega(0) + T_L(0)]/J$

Iteração 2:
* $ k = 1, i_a(1) = $ calculado na iteração anterior
* $ i_a(2) = i_a(1) + h . [V_a(1) - R_a.i_a(1) - K_\phi.\omega(1)]/L_a$ 
* $ \omega(2) = \omega(1) + h.[K.i_a(1) + B.\omega(1) + T_L(1)] / J$

````python
import numpy as np
import matplotlib.pyplot as plt

def f(t, y):
    """The differential equation dy/dt = f(t, y)"""
    return t + y

def euler_method(f, t0, y0, h, tn):
    """
    f : Function defining the ODE
    t0, y0: Initial conditions
    h : Step size
    tn: Final time value
    """
    t_values = np.arange(t0, tn + h, h)
    y_values = np.zeros(len(t_values))
    y_values[0] = y0
    
    for i in range(0, len(t_values) - 1):
        # Update rule: y_next = y_curr + h * f(t_curr, y_curr)
        y_values[i+1] = y_values[i] + h * f(t_values[i], y_values[i])
        
    return t_values, y_values

# Parameters
t0, y0 = 0, 1
h = 0.1
tn = 2

# Solve
t, y = euler_method(f, t0, y0, h, tn)

# Plotting results
plt.plot(t, y, 'bo-', label='Euler Approximation')
plt.xlabel('t')
plt.ylabel('y(t)')
plt.legend()
plt.title("Euler's Method Solution")
plt.show()
````
