# __Diagrama de Bode__

### 4.1 Introdução

Diferente das abordagens de análise de sistemas anteriores, os **Diagramas de Bode** analisam sistemas no __domínio da Frequência__.  Assim, ao invés de entradas impulso e degrau unitários, as entradas são geralmente __senoidais__ e busca-se como um sistema se comporta conforme varia-se a frequência desse sinal.

Em geral, usamos a Transformada de Laplace da entrada senoidal no domínio do tempo e substituimos a variável $(s)$ por $(jw)$ para obtermos a Transformada de Fourier do sinal de entrada.

No tempo: 
$x(t) = A_p.sen(wt)$

Na frequênca:
$X(j\omega) = \dfrac{A_p.\omega}{(j\omega)^2 + \omega^2}$

Os parâmetros analisados nos Diagramas de Bode são o **ganho** do sinal e o **defasagem** do sinal de saída em relação ao de entrada. Por isso os Diagramas de Bode são representados por dois gráficos, um que relaciona magnitude com frequência e outro que relaciona a fase com a frequência.

Nos Diagramas de Bode o eixo horizontal representa a frequência em escala logarítmica e o eixo vertical representa o ganho em escala linear (decibel, dB).

### 4.2 Criação dos Diagramas de Bode

Os diagramas podem ser construídos de duas formas: Computacionalmente, usando vários pontos de frequência e calculando individualmente a magnitude e a fase, ou usando assíntotas.

A segunda forma, usando assíntotas, podemos facilmente extraí-las a partir do conhecimento de como cada termo da Função de Transferência (F.T.) altera o sistema e fazendo a soma de todas essas alterações. Apesar de se ter um erro quando se traça a curva assíntotica, o mesmo é no máximo 3 dB como mostra o gráfico.

### 4.3 Análise Assintótica dos Termos da F.T.

#### 4.3.1 Ganho Proporcional

Um ganho proporcional gera uma alteração no módulo amplificando ou atenuando um sinal mas não causa qualquer alteração na fase do sinal. Caso esse ganho seja maior que um teremos um ganho em dB positivo, enquanto se o ganho estiver entre 0 e 1, teremos um ganho em dB negativo.  

#### 4.3.2 Polo ou Zero na Origem

A presença de um polo ou zero da F.T. na origem, termo $s$ ou termo $1/s$, trazem um ganho ou perda linear de 20 dB, respectivamente, por cada década de frequência percorrida.  
Enquanto no gráfico da fase um polo na origem gera um atraso de 90° para qualquer frequência, enquanto que o zero na origem gera um avanço de 90°.

#### 4.3.3 Polo ou Zeros Reais

A presença de polos ou zeros no eixo real da F.T. fora da origem, $(s+n)$ ou $1/(s+n)$, ou normalizando: $(T.s+1)$ e $1/(T.s+1)$ mantém a magnitude sem alteração até a frequência que corresponde $1/T$ (dB), onde se começa a subir ou descer 20 dB caso o elemento deja um zero ou um polo, respectivamente.  
Em termos de defasamento, na frequência 1/T teremos uma variação de $\pm45°$, e nas décadas anteriores permanece em 0°, enquanto que nas décadas seguintes teremos $\pm90°$.

#### 4.3.4 Polos ou Zeros Complexos Conjugados

Os polos ou zeros complexos conjugados, que podem ser escritos como: 
* Zeros: $(as^2 + bs + c)$ ou $(s^2 + 2\zeta w_n + w_n^2)$, ou normalizado: $(\dfrac{s^2}{w_n^2}+ \dfrac{2\zeta }{w_n}s + 1)$ 
* Polos:  $\dfrac{1}{(as^2 + bs + c)}$ ou $\dfrac{1}{(s^2 + 2\zeta w_n + w_n^2)}$, ou normalizado: $\dfrac{1}{(\dfrac{s^2}{w_n^2}+ \dfrac{2\zeta }{w_n}s + 1)}$  
Estes se comportam na magnitude permanecendo sem alteração até chegar na frequência $w_n$, e a partir dela cresce ou decresce 40 dB/déc enquanto que em termos de fase, na frequência $w_n$ teremos $\pm90°$, e nas décadas anteriores em 0° enquanto que nas décadas posteriores em $\pm180°$.
