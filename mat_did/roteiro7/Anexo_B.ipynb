# Python: __Métodos Dunders__

## 1. Introdução

Vamos mostrar como implementar os métodos especiais da linguagem Python, os __Dunder__'s, assim chamados por terem seus nomes iniciados e terminados por duplos sublinhados, em inglês, _Double Underscore_:  
* __init__: usado na construção/criação de um novo objeto, ou seja, esse método define o que se quer que aconteça na instanciação da classe em questão;
* __str__: usado na exibição de uma string que represente um objeto da classe em questão.

Eles podem ser usados para definir comportamentos personalizados para operadores e funções diversas, tais como: impressão, comparação e manipulação de objetos em Python.

```python
# Métodos Dunder's, ou métodos mágicos, em Python
# Exemplo do método __init__
class Retangulo:
    def __init__(self, x, y):
        self.x = x
        self.y = y

# Criando um objeto 'Retangulo' com lados 5 e 12
ret1 = Retangulo(5, 12)
print(ret1, ret1.x, ret1.y)
```

## 2. Tipos Básicos (nativos) em Python

Antes de falarmos dos tipos de dados básicos em Python, precisamos ter em mente que na linguagem Python **tudo** que for criado num programa Python será considerado um **objeto**.

Podemos usar a função `type()` para mostrar o tipo de dados de qualquer objeto em Python.

#### Exemplo

```python
str1 = "Oi "         # essa linha cria um objeto String, chamado 'str1' 
str2 = "Pessoal!"    # essa linha cria outro objeto String, chamado 'str2' 
print(5)             # essa linha cria uma constante do tipo Inteiro como argumento da função 'print'
                     #  e a própria função é também um objeto do tipo Função
def minha_funcao():
    pass

print(type(str1), type(str2), type(5), type(print))
print(f"'minha_funcao()' é uma instância da {type(minha_funcao)}")
```

* **String**  
Porque em Python podemos somar duas strings, como mostra o seguinte exemplo... A razão porque podemos fazer isso é que esses objetos são de um mesmo tipo de dados e que a ling. Python implementa esse comportamento de concatenação de sequência de caracteres com um método _Dunder_ chamado `__add__()`, definido na classe `'str'`.

```python
print(str1 + str2)
str3 = str1 + str2
str4 = str1.__add__(str2)
print(str3, str4)
```

Mais um método mágico da classe `'str'` é um método que retorna o comprimento da string: `__len__()`.

```python
print(len(str3), str3.__len__())
```

__Conclusão__: todo objeto tem uma classe da qual ele é instanciado, e que define o comportamento dele, seja na operação de soma ativada pelo operador `+` (`__add__()`), como na obtenção do seu comprimento (`__len__()`) ou de como ele será exibido numa impressão (`__str()__`).

## 3. Criando seus Próprios Métodos Dunder's

Considere o exemplo seguinte:

```python
class Contador:
   def __init__(self):
      self.valor = 1

   def avanca(self):
      self.valor += 1
    
   def recua(self):
      self.valor -= 1

cont1 = Contador()
cont2 = Contador()
cont1.avanca()
cont2.avanca()
print(type(cont1), type(cont2))
print(cont1, cont2)
```

Observe que ao tentarmos mostrar os valores dos dois objetos da classe `Contador` obtivemos referências da localização em memória de cada um dos objetos... Em geral, o usuário está querendo uma informação mais pragmática dos objetos, por exemplo, nesse caso poderia ser o próprio valor de contagem de cada objeto. Isso pode ser providenciado com a implementação do *dunder* `__str__()`, como mostramos no seguinte código.

```python
class Contador:
    def __init__(self):
        self.valor = 1

    def __str__(self):
        return f"Contagem: {self.valor}"

    def avanca(self):
        self.valor += 1
    
    def recua(self):
        self.valor -= 1

cont1 = Contador()
cont2 = Contador()
cont1.avanca()
cont2.avanca()
print(type(cont1), type(cont2))
print(cont1, cont2)
print(cont1 + cont2)
```

O erro relatado na célula anterior, `TypeError: tipo(s) de operandos não permitidos para '+': 'Contador' e 'Contador'`, ocorreu porque a classe `Contador` não implementou a operação de soma com um *dunder*, `__add__()`.  

Vamos corrigir essa questão:

```python
class Contador:
    def __init__(self):
        self.valor = 1

    def __str__(self):
        return f"Contagem: {self.valor}"

    def __add__(self, segParam):
        return self.valor + segParam.valor

    def avanca(self):
        self.valor += 1
    
    def recua(self):
        self.valor -= 1

cont1 = Contador()
cont2 = Contador()
cont1.avanca()
cont2.avanca()
print(type(cont1), type(cont2))
print(cont1, cont2)
print("Soma das contagens:", cont1 + cont2)
```
