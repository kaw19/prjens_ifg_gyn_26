# __Diagrama de Bode__

### 4.1 Introdução

Diferente das abordagens anteriores para análise de sistemas, os **Diagramas de Bode** analisam sistemas no __domínio da Frequência__.  Assim, ao invés de entradas com sinais impulso e degrau unitários, as entradas são geralmente sinais __senoidais__, usadas para se avaliar como o sistema se comporta variando a frequência desse sinal em determinada faixa de valores.  

Os **Diagramas de Bode** são representações gráficas da **Resposta em Frequência** de um sistema: __Magnitude__ (em dB) e __Fase__ (em graus), ambos em função da frequência angular, ω (rad/s), em escala logarítmica.

Em geral, usamos a Transformada de Fourier da entrada senoidal no domínio do tempo, obtendo o resultado no domínio da frequência angular, $(jw)$.

O sinal de entrada no tempo:  $x(t) = A_p.sen(wt)$

Usando a Transformada de Fourier, obtemos no domínio da frequênca: $X(j\omega) = \dfrac{A_p.\omega}{(j\omega)^2 + \omega^2}$

Os parâmetros analisados nos Diagramas de Bode são o **ganho** do sinal de saída em relação ao sinal de entrada, e a **defasagem** do sinal de saída em relação ao de entrada. Por isso os Diagramas de Bode são representados por dois gráficos, um que relaciona magnitude com frequência e outro que relaciona a fase com a frequência.

Os **Diagramas de Bode** são gráficos semi-log, onde o eixo horizontal representa a frequência (em escala logarítmica) e o eixo vertical representa o ganho (em escala linear, porém usando uma unidade logarítmica, o decibel - dB) ou o ângulo da fase (em escala linear, em graus).

### 4.2 Criação dos Diagramas de Bode

Os diagramas podem ser construídos basicamente de duas formas: 
* Computacionalmente: usando vários pontos de frequência e calculando individualmente a magnitude e a fase para cada ponto;
* Retas Assíntotas: análise individual dos termos da Resposta em Frequência do sistema e traçado aproximado das curvas correspondentes aos diversos termos.

A segunda forma, usando as assíntotas, pode ser facilmente esboçada a partir do conhecimento de como cada termo da R.F. altera o sistema, e depois somando todas essas alterações. Apesar de se ter um erro ao se traçar as assíntotas, o mesmo é no máximo 3 dB como mostra a figura seguinte.

* Erro em dB da assíntota, para amortecimento crítico ($\zeta = 1$):

### 4.3 Análise Assintótica dos Termos da F.T.

Para desenhar os diagramas de forma aproximada, decompomos a F.T. em fatores elementares, rescrevendo-a conforme formato padrão de Bode, por exemplo, para termos de primeira ordem:
$$G(s)= K.\dfrac{1+T_z.s}{1+T_p.s}​$$

#### 4.3.1 Ganho Constante (K)

Um ganho constante, $K$, gera uma alteração na magnitude, amplificando ou atenuando a magnitude da R.F., $\lvert {G(j\omega)} \rvert$, mas não causa qualquer alteração na fase da R.F., $\angle G(j\omega)$. Caso $K>1$, teremos ganho positivo em dB, enquanto $0 < K < 1$, teremos um ganho negativo em dB.  

#### 4.3.2 Polos e Zeros na Origem ($s^{\pm n}$)

A presença de um polo ou zero da F.T. na origem, termo $s$ ou $1/s$, gera ganho/perda linear de 20 dB por década de frequência.  
No gráfico de fase, um polo na origem gera um atraso de 90° para toda frequência, e um zero na origem gera um avanço de 90° para toda frequência.

* Magnitude: reta com inclinação de −20.n dB/década (para polos) ou +20.n dB/década (para zeros), cruzando 0 dB em ω = 1 rad/s.
* Fase: Constante em $−90^∘.n$ (para polos) ou $+90^∘.n$ (zeros).

#### 4.3.3 Polos ou Zeros Reais e Distintos ($1+T.s$)

A frequência de corte (ou de quebra) nesses casos será $\omega_c = 1/T$.  
* Magnitude  
  * Para $\omega < \omega_c$: reta paralela ao eixo horizontal, em 0 dB.  
  * Para $\omega > \omega_c$: reta com inclinação de −20 dB/década (para polo) ou +20 dB/década (para zero).  
* Fase:
  * Aproximação assintótica: começa a mudar uma década antes ($\omega_c/10$) e estabiliza uma década depois ($10.\omega_c)$. Na frequência $\omega_c$, a fase é de $−45^∘$ (para polo) ou $−45^∘$ (para zero).

#### 4.3.4 Polos ou Zeros Complexos Conjugados

Os polos ou zeros complexos conjugados, que podem ser escritos como: 
* Zeros de segunda ordem:  $s^2 + 2\zeta w_n + w_n^2$  
* Polos de segunda ordem:  $\dfrac{1}{s^2 + 2\zeta w_n + w_n^2}$   
Estes se comportam na magnitude permanecendo sem alteração até chegar na frequência $w_n$, e a partir dela cresce ou decresce 40 dB/déc enquanto que em termos de fase, na frequência $w_n$ teremos $\pm90°$, e nas décadas anteriores em 0° enquanto que nas décadas posteriores em $\pm180°$.

### 4.4 Exemplo, Esboço e Simulação
#### 4.4.1 Exemplo: Controle de Velocidade de Motor DC
Vamos modelar um Motor DC simplificado, onde a entrada é a tensão de armadura ($V_a$) e a saída é a velocidade angular ($\omega_m$). A F.T. clássica de primeira ordem (desprezando a indutância da armadura para simplificar) é: $$ G(s) = \dfrac{\Omega_m(s)}{V_a(s)} = \dfrac{K_m}{T_m.s+1}$$

Supondo os seguintes parâmetros para o Motor DC:
- Ganho do motor: $K_m = 10$ (rad/s)/V
- Constante de tempo mecânico: $T_m = 0,1$ s

F.T. do Motor DC: $$ G(s) =\dfrac{10}{0,1.s+1}$$

#### 4.4.2 Esboco Assintótico
1. Identificação dos termos de $G(s)$: temos um ganho $K=10$ e um polo real com $T_m = 0,1 \Rightarrow \omega_c = 1/0,1 = 10$ rad/s
1. Desenho da Magnitude:
    * De $\omega = 0,1$ até $\omega =10$ rad/s: o ganho será de $20.log_{10} 10 = 20$ dB, portanto gerando uma reta horizontal em 20 dB.
    * A partir de $\omega = 10$ rad/s: o polo gera uma reta com inclinação de −20 dB/década. (Em $\omega = 100$, a reta estará em 0 dB).
3. Desenho da Fase:  
    * De $\omega = 0,1$ até $\omega = 1$ rad/s (10.$\omega_c$ antes): fase em 0$^∘$.
    * De $\omega = 1$ até $\omega = 100$ rad/s: reta inclinada caindo até −90$^∘$.
    * Na frequência de corte ($\omega = 10 rad/s): a fase é exatamente −45$^∘$.
    * Acima de $\omega = 100$ rad/s: fase estabiliza em −90$^∘$.

#### 4.4.3 Simulação    

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ct

# Definindo os parâmetros do Motor DC
K_m = 10       # Ganho do motor
tau_m = 0.1    # Constante de tempo (s) -> Frequência de corte = 10 rad/s

# Definindo a Função de Transferência: G(s) = 10 / (0.1s + 1)
num = [K_m]; den = [tau_m, 1]; G_motor = ct.TransferFunction(num, den)

# Gerando o Diagrama de Bode
# dB=True exibe a magnitude em decibéis, deg=True exibe a fase em graus
plt.figure(figsize=(10, 5))
mag, fase, omega = ct.bode_plot(G_motor, dB=True, deg=True, omega_limits=(0.1, 1000), plot=True)

# Customizando o gráfico para destacar a frequência de corte
axes = plt.gcf().axes # Captura os eixos de magnitude (0) e fase (1)

# Linha vertical na frequência de corte (10 rad/s)
axes[0].axvline(10, color='r', linestyle='--', label=r'$\omega_c = 10$ rad/s')
axes[0].legend()
axes[1].axvline(10, color='r', linestyle='--', linewidth=3)

# Linha horizontal no ganho DC (20 dB)
axes[0].axhline(20, color='g', linestyle=':', linewidth=3, label='Ganho DC (20 dB)')

plt.suptitle('Diagramas de Bode - Controle de Velocidade do Motor DC')
plt.show()
```

<u>Intuição da engenharia</u>: O motor funciona como um filtro passa-baixas. Ele responde bem a comandos de velocidade constantes ou de baixa frequência, mas se tentarmos variar a tensão de entrada (armadura) rápido demais (alta frequência), o motor não consegue acompanhar devido à sua inércia ($\tau_m$).

#### 4.4.4 Adicionando Controlador PI ao Sistema
O controlador PI tem a seguinte F.T.:
$$C(s) = K_p + \dfrac{K_i}{s}= \dfrac{K_p.s+K_i}{s} = K_p.\dfrac{s+\frac{K_i}{K_p}}{s} $$

Repare que o controlador PI adiciona dois termos ao sistema:
1. Um polo na origem ($1/s$): garante erro zero em regime permanente para entrada degrau (o motor vai atingir exatamente a velocidade desejada).  
1. Um zero real ($−K_i/K_p$): usado para compensar o atraso de fase gerado pelo polo do motor.

Vamos escolher os ganhos do controlador PI de forma estratégica para cancelar o polo do motor (técnica de cancelamento polo por zero):
* Desejamos que o zero do PI aconteça exatamente na frequência de corte do motor (ω=10 rad/s).
* Portanto: $K_p/K_i = 10 ⟹ K_i = 10.K_p$.
* Vamos definir $K_p = 0,5$, logo $K_i = 5$.

A F.T. de malha aberta compensada, $G_{ma}(s)=C(s)⋅G(s)$ fica:  
$$G_ma(s)=\frac{0,5(s+10)}{s}.\frac{10}{0,1s + 1}=\frac{0,5 . 10(s+10)}{s.0,1(s+10)}=\frac{5}{0,1s}=\frac{50}{s}$$

**Resultado no Esboço Assintótico**:  
Ao cancelar o polo do motor com o zero do controlador PI, o sistema resultante se comporta puramente como um Integrador com Ganho (50/s):
* **Magnitude**: Uma linha reta que desce infinitamente com inclinação de −20 dB/década. Ela cruza o eixo de 0 dB exatamente em ω = 50 rad/s.
* **Fase**: Uma linha horizontal constante em $−90^∘$.  O sistema se tornou extremamente estável (margem de fase de $90^∘$).

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ct

# Sistema Original (Motor DC)
K_m = 10.; tau_m = 0.1
G_motor = ct.TransferFunction([K_m], [tau_m, 1])

# Controlador PI
Kp = 0.5; Ki = 5.0
C_pi = ct.TransferFunction([Kp, Ki], [1, 0]) # Kp*s + Ki / s

# Sistema em Malha Aberta Compensado (C * G)
G_malha_aberta = ct.series(C_pi, G_motor)

# Plotando os dois diagramas para comparação
plt.figure(figsize=(10, 5))

# Bode do Motor Original (Linha Azul)
mag_m, fase_m, omega_m = ct.bode_plot(G_motor, dB=True, deg=True, label='Motor', color='blue', plot=True)

# Bode do Sistema com PI (Linha Laranja)
mag_pi, fase_pi, omega_pi = ct.bode_plot(G_malha_aberta, dB=True, deg=True, label='Motor + Controlador PI', color='orange', plot=True)

# Customização dos Gráficos
axes = plt.gcf().axes

# Gráfico de Magnitude (Topo)
axes[0].axvline(10, color='r', linestyle='--', label=r'$\omega_c$ do Motor (10 rad/s)')
axes[0].legend()
axes[0].set_title('Resposta em Magnitude')

# Gráfico de Fase (Baixo)
axes[1].axvline(10, color='r', linestyle='--')
axes[1].set_title('Resposta em Fase')

plt.suptitle('Comparativo de Bode: Efeito do Controlador PI no Motor DC', fontsize=14)
plt.tight_layout()
plt.show()
```

#### 4.5 Malha Fechada e Resposta ao Degrau

Ao fechar a malha do controlador PI e planta, vamos ver como fica a resposta a um degrau com velocidade específica.

A função de transferência em malha fechada $T(s)$ (considerando realimentação unitária negativa) é dada por:
$$T(s)= \dfrac{G_{ma}(s)}{1+G_{ma}(s)}$$
 
Como vimos no item anterior, a malha aberta compensada foi simplificada, como $G_{ma}(s) = 50/s$. Substituindo na equação da malha fechada:  
$$T(s)= \dfrac{50/s}{1+50/s} = \dfrac{50}{s+50}$$

**Resultados**  
* Redução da ordem do Sistema Controlado: O sistema em malha fechada se tornou um sistema de primeira ordem puro, sem _overshoot_ (sem oscilação/sobrepasso).  
* Nova Constante de Tempo: A nova frequência de corte em malha fechada é $\omega = 50\text{ rad/s}$. Isso significa que a nova constante de tempo do sistema é $\tau_{mf} = \frac{1}{50} = 0,02\text{ s}$.  
* Velocidade de Resposta: O motor original demorava cerca de $4\tau_m = 0,4\text{ s}$ para estabilizar. Com o controlador PI em malha fechada, ele estabiliza em $4\tau_{mf} = 0,08\text{ s}$. O sistema ficou 5 vezes mais rápido!  
* Erro Zero: O ganho DC em malha fechada (substituindo $s=0$ na F.T. de malha fechada) é $T(0) = \frac{50}{50} = 1$. Erro zero em regime permanente está garantido!  

Vamos atualizar o código para plotar os **Diagramas de Bode** de Malha Aberta ao lado da **Resposta ao Degrau** em Malha Fechada.

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ct

# Definição do Motor e Controlador PI (mesmos dados das simulações anteriores)
K_m = 10; tau_m = 0.1
G_motor = ct.TransferFunction([K_m], [tau_m, 1])  # F.T. do Motor DC

Kp = 0.5; Ki = 5.0; 
C_pi = ct.TransferFunction([Kp, Ki], [1, 0])      # F.T. do Controlador PI

G_malha_aberta = ct.series(C_pi, G_motor)         # F.T. da Malha Aberta Compensada

G_malha_fechada = ct.feedback(G_malha_aberta, 1)  # F.T. da Malha Fechada com realimentação unitária negativa: T(s) = Gma / (1 + Gma)

# Resposta ao Degrau no Tempo
# Simulando a resposta a um degrau unitário (entrada de referência: veloc. angular de 1 rad/s)
t, y = ct.step_response(G_malha_fechada, T=np.linspace(0, 0.2, 500))

# Gráficos lado a lado para comparação
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.plot(t, y, 'g-', linewidth=2, label='Velocidade do Motor ($\omega_m$)')
ax1.axhline(1.0, color='r', linestyle='--', label='Referência (SetPoint = 1)')
ax1.set_title('Resposta ao Degrau (Domínio do Tempo)'); ax1.set_xlabel('Tempo (segundos)'); 
ax1.set_ylabel('Velocidade (rad/s)'); ax1.grid(True); ax1.legend()

# Destacando o tempo de acomodação aproximado (4 * tau = 0.08 s)
ax1.axvline(0.08, color='purple', linestyle=':', label='Tempo de Estabilização (~0.08s)'); ax1.legend()

# Diagramas de Bode do Sistema em Malha Aberta
mag, fase, omega = ct.bode_plot(G_malha_aberta, omega_limits=(1, 1000), Plot=False)
mag_db = 20 * np.log10(mag)                       # Convertendo a magnitude para dB
phase_deg = np.degrees(fase)                      # Convertendo fase para graus

ax2.semilogx(omega, mag_db, 'orange', linewidth=2, label='Malha Aberta com PI')
ax2.axvline(50, color='purple', linestyle='--', label='Frequência de Cruzamento ($\omega_g = 50$ rad/s)')
ax2.set_title('Diagrama de Bode de Magnitude (Domínio da Frequência)')
ax2.set_xlabel('Frequência (rad/s)'); ax2.set_ylabel('Magnitude (dB)')
ax2.grid(True, which="both", ls="-"); ax2.legend()

plt.suptitle('Conexão Frequência-Tempo: Projeto de Controle do Motor DC', fontsize=14)
plt.tight_layout(); plt.show()
```
