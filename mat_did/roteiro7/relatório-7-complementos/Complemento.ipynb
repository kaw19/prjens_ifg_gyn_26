# Diagrama de Bode

## 3.1. Introdução

#### Diferente dos sistemas vistos anteriormente, o diagrama de Bode analiza sistemas no __domínio da frequência__. Assim as entradas são geralmente __senoidais__ e  que se analiza é como um sistema se comporta conforme se varia a frequência desse sinal.

#### Por isso, se usa a transformada de Laplace para transformar uma entrada senoidal no domínio do tempo em uma no domínio da frequência.

### No tempo: 
$\large x(t) = x_osen(wt)$

### Na frequênca:
$\large X(S) = \dfrac{aw}{s^2 + w^2}$


#### Os parâmetros que são analizados nesse diagrama são a amplificação e atenuação do sinal, denominada __módulo__ ou __Magnitude__ quando transformada para forma logarítimica e o defasamento do sinal, denominado __Fase__ em relação a variação de frequência. Por isso o diagrama de Bode é representado por dois gráficos, um que relaciona magnitude com frequência e outro que relaciona a fase com a frequência.

#### Tanto a frequência como a magnitude são medidos no diagrama no formato logarítimo (decibel).

## Criação do diagrama de Bode

#### Os diagramas podem ser construídos de duas formas: Computacionalmente, usando varios pontos de frequência e calculando manualmente a magnitude e a fase ou usando assíntotas.

#### Manualmente, se é mais conveniente construir usando assintotas que podem ser facilmente extraídas a partir do conhecimento de como cada termo de uma função de transferência altera o sistema e somando todas essas alterações. Apesar de se ter um erro quando se traça a curva assíntotica, o mesmo é no máximo 3dB como mostra o gráfico.

### 2.1 Ganho proporcional

#### Um ganho proporcional proporciona uma alteração no módulo amplificando ou atenuando um sinal mas não causa qualquer alteração na fase do sinal. Caso esse ganho seja > 1 se terá um ganho em dB positivo enquanto se for entre 0 e 1 terá um ganho em dB negativo.  

### 2.2 Polo ou zero na origem

#### A presença de um polo ou zero na origem que é: $s$ ou $\dfrac{1}{s}$, trazem um ganho ou perda linear de 20 dB por cada década de frequência que se aumenta.

#### Enquanto no gráfico da fase um polo na origem o trás para -90° para qualquer frequência enquanto o zero na origem o trás para +90°

### 2.3 Polo ou zeros reais

#### A presença de polos ou zeros no eixo real fora da origem que se escreve: $(s+n)$ ou $\dfrac{1}{(s+n)}$, ou normalizando: $(Ts+1)$ e $\dfrac{1}{(Ts+1)}$ matém a magniyude sem alteração até a frequência que corresponde 1/T (dB) aonde se começa a subir ou descer 20 dB caso o elemento deja um zero ou um polo respectivamente.
#### Já na fase, em 1/T ele se encontra em $\pm45°$ e na década anterior e para trás permanece em 0° enquanto na década seguinte e afrente se encontra em $\pm90°$

### 2.4 Polos ou zeros complexos conjugados

#### Os polos ou zeros complexos conjugados, que podem ser escritos: 
#### os zeros: $(as^2 + bs + c)$ ou $(s^2 + 2\zeta w_n + w_n^2)$ que quando normalizado fica: $(\dfrac{s^2}{w_n^2}+ \dfrac{2\zeta }{w_n}s + 1)$ 
#### enquanto os polos:  $\dfrac{1}{(as^2 + bs + c)}$ ou $\dfrac{1}{(s^2 + 2\zeta w_n + w_n^2)}$ que quando normalizado fica: $\dfrac{1}{(\dfrac{s^2}{w_n^2}+ \dfrac{2\zeta }{w_n}s + 1)}$
#### Estes se comportam no modulo permanecendo sem alteração até chegar na frequência $w_n$ e a partir dela se cresce ou decresce 40 dB por década
#### enquanto na fase no $w_n$ ele se encontra em $\pm90°$ e na década anterior e as atrás em 0° enquanto nas posteriores em $\pm180°$
