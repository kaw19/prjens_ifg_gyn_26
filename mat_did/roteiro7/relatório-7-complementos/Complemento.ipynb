# Diagrama de Bode

## 1.Introdução

#### Diferente dos sistemas vistos anteriormente, o diagrama de Bode analiza sistemas no __domínio da frequência__. Assim as entradas são geralmente __senoidais__ e  que se analiza é como um sistema se comporta conforme se varia a frequência desse sinal.

#### Por isso, se usa a transformada de Laplace para transformar uma entrada senoidal no domínio do tempo em uma no domínio da frequência.

### Na frequênca: <img src="fig.3.jpg" alt="Plano s mostrando o lugar das raízes de um sistema de segunda ordem" heigth="250" style="display: block; margin: 0 auto;">

#### Os parâmetros que são analizados nesse diagrama são a amplificação e atenuação do sinal, denominada __módulo__ ou __Magnitude__ quando transformada para forma logarítimica e o defasamento do sinal, denominado __Fase__ em relação a variação de frequência. Por isso o diagrama de Bode é representado por dois gráficos, um que relaciona magnitude com frequência e outro que relaciona a fase com a frequência.

#### Tanto a frequência como a magnitude são medidos no diagrama no formato logarítimo (decibel).

<img src="fig.1.jpg" alt="Plano s mostrando o lugar das raízes de um sistema de segunda ordem" heigth="250" style="display: block; margin: 0 auto;">

#### Esse diagrama também pode ser usado para verificar a instabilidade de um sistema que depende da __margem de fase__ e __margem de ganho__ de um sistema.

## Criação do diagrama de Bode

#### Os diagramas podem ser construídos de duas formas: Computacionalmente, usando varios pontos de frequência e calculando manualmente a magnitude e a fase ou usando assíntotas.

#### Manualmente, se é mais conveniente construir usando assintotas que podem ser facilmente extraídas a partir do conhecimento de como cada termo de uma função de transferência altera o sistema e somando todas essas alterações.
