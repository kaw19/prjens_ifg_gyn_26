#### 1.1 Classes e Objetos
Para se criar uma classe em Python usamos a palavra-chave `class` seguida pelo nome da classe.  

Sintaxe:  
``` python
    class NomeDaClasse:
```

E para criar objeto(s) de uma classe em Python usamos a seguinte sintaxe:

Sintaxe:
``` python
    nomeDoObjeto = NomeDaClasse(parâmetros)
```

##### Camel Case
O _camel case_ é um estilo de escrita de nomes compostos, sem espaços em branco, onde a primeira letra de cada palavra em uma frase é maiúscula. A tradução literal é "caixa camelo", pois as letras maiúsculas no meio do texto lembram as corcovas (corcundas) nas costas de um camelo.

<img src='./camelcase.jfif' width=300>

O nome da classe, convencionalmente, segue a nomenclatura **_camel case_** que aglutina palavras sem usar espaços em branco, sendo a primeira palavra com letras minúsculas e as demais palavras com a primeira letra em maiúscula e as demais letras em minúsculas (ex.: **meuPrimeiroExemplo**). Excepcionalmente, para nominar as classes em Python, devemos usar a primeira palavra também iniciando com letra maiúscula (ex.: **FornoMicroOndas**).

A convenção tem o objetivo de melhorar a legibilidade em cenários onde não é possível usar espaços. Ela se divide principalmente em duas variações:
* ***lowerCamel Case***: A primeira letra é minúscula e as demais são maiúsculas (ex.: **nomeDoUsuario**).  
* ***UpperCamel Case*** (_Pascal Case_): Todas as palavras começam com letra maiúscula (ex.: NomeDoUsuario). Essa variação geralmente é recomendada para a criação de nomes de classes em linguagens orientadas a objetos.

### <span style="color: #0040ff;" id="Exemplo">2. Exemplo

#### Forno de Micro-ondas

##### Definição:
O forno de micro-ondas é um eletrodoméstico que aquece e cozinha alimentos utilizando ondas eletromagnéticas de alta frequência que agitam as moléculas de água, gordura e açúcar presentes nos alimentos. Essas moléculas vibram milhões de vezes por segundo, gerando calor por fricção, ou seja, o calor é gerado diretamente no interior do alimento, tornando o processo de aquecimento muito rápido.  

##### Exemplo:
Vamos supor que somos um fabricante de fornos de micro-ondas. Primeira coisa a se fazer para produzir fornos de micro-ondas é projetar um ou mais modelos para serem manufaturados, ou seja, precisamos criar uma __planta__ contendo desenhos mecânicos, elétricos e eletrônicos detalhados (**classe**) para em seguida construir uma linha de produção em série dos fornos (**objetos**).

````python
# Definindo a planta baixa da casa/forno, a Classe
class FornoMicroOndas:
    pass

# Instanciando um objeto da classe FornoMicroOndas (criando um objeto dessa classe)
philco = FornoMicroOndas()      # o método construtor de objetos, sempre usa o mesmo nome dado à classe, seguido de parênteses
print(philco)                   # ID único apontando (endereço) para a área de memória ocupada pelo objeto (instância de FornoMicroOndas)
````

O exemplo anterior define uma classe (uma planta baixa da construção, no caso de forno de micro-ondas) e cria um objeto da classe **FornoMicroOndas**, e nesse caso, chamado **forno**. Observe que a planta baixa não ocupa espaço de memória, porém os objetos criados a partir da classe ocupam espaço de memória para armazenar os atributos (características) de cada objeto.  

A classe definida na célula anterior não possui nenhuma utilidade prática ainda, pois não armazena nenhuma informação e nem permite a execução de qualquer funcionalidade.

##### Inicializadores
São métodos que permitem a criação de objetos com informações específicas que individualizam cada forno de micro-ondas, tais como: fabricante, categoria de consumo de energia, capacidade volumétrica (litros) etc.  

O método inicializador em Python é um método Dunder (_double underscores_) chamado **\_\_init\_\_()**. Toda vez que um objeto é criado esse método é disparado automaticamente.

##### Atributos
Vamos acrescentar os seguintes atributos à classe **FornoMicroOndas**:
- fabricante
- categoria de consumo de energia
- capacidade volumétrica

````python
# Definindo a Classe
class FornoMicroOndas:
    # O método __init__ é o construtor do objeto, e define os atributos que cada objeto terá
    def __init__(self, fabricante, catConsumo, volume):
        self.fabricante = fabricante      # atributo
        self.catConsumo = catConsumo      # atributo que indica a categoria de consumo de energia
        self.volume = volume              # atributo medido em litro

# Criando um forno (objeto)
forno1 = FornoMicroOndas("Philco","B",22)
print(f"Forno da marca {forno1.fabricante}, da categoria de consumo {forno1.catConsumo} e com capacidade de {forno1.volume} L.")
````

O parâmetro `self` passado ao método `__init__()` representa a instância da classe e que apenas garante que esses dados fiquem na instância certa ou objeto correto, portanto, é como se o objeto **forno1** ou outro objeto qualquer, fosse usado no método `__init__()`:  

``` python
def __init__(self, fabricante, catConsumo, volume):
        self.fabricante = fabricante    -->   forno1.fabricante = fabricante     # atributo
        self.catConsumo = catConsumo    -->   forno1.catConsumo = catConsumo     # atributo
        self.volume = volume            -->   forno1.volume = volume             # atributo
```
O parâmetro `self` em si pode ter qualquer nome, tal como `this` que normalmente é usado em Java, ou em C++, para designar o objeto em questão, porém `self` já se tornou uma convenção em Python.


````python
# Criando outro forno (mais um objeto)
forno2 = FornoMicroOndas("Electrolux","C",40)
print(f"Forno da marca {forno2.fabricante}, da categoria de consumo {forno2.catConsumo} e capacidade de {forno2.volume} L.")
````

##### Métodos
No exemplo da classe **FornoMicroOndas** até esse momento, cada objeto armazena suas características ou atributos, mas não realiza nenhuma ação com esses dados armazenados... Comecemos pelas funções `ligar()` e `desligar()` o forno de micro-ondas, na qual um novo atributo manterá o estado da instância da classe: ligado ou desligado; e encerramos essa classe com um método para `aquecer()` alimentos.

````python
# Definindo a Classe
class FornoMicroOndas:
    # O método __init__ é o construtor do objeto, e define os atributos que cada objeto terá
    def __init__(self, fabricante, catConsumo, volume):
        self.fabricante = fabricante      # atributo
        self.catConsumo = catConsumo      # atributo
        self.volume = volume              # atributo
        self.ligado = False               # atributo booleano que indica se o forno está ligado ou desligado

    # Definindo métodos (comportamento)
    def ligar(self):
        if self.ligado:
            print(f"Forno micro-ondas ({self.fabricante}) já está ligado!")
        else: 
            self.ligado = True
            print(f"Forno micro-ondas ({self.fabricante}) agora está ligado.")

    def desligar(self):
        if self.ligado:
            self.ligado = False
            print(f"Forno micro-ondas ({self.fabricante}) agora está desligado.")
        else: 
            print(f"Forno micro-ondas ({self.fabricante}) já está desligado!")

    def aquecer(self, segundos):
        if self.ligado:
            print(f"Aquecendo ({self.fabricante}) por {segundos} seg.")
        else:
            print(f"Vc precisa ligar o forno micro-ondas ({self.fabricante}) primeiro!")

# Criando uma instância (objeto) da classe FornoMicroOndas: forno1 
forno1 = FornoMicroOndas("Philco","B",22)

# Usando as funcionalidades do objeto criado: forno1
forno1.ligar()                                # acionamento do método `ligar()` usando a notação de ponto (dot)
forno1.ligar()                                # ligando duas vezes... só para ver a mensagem
forno1.aquecer(30)                            # aquecer alimento por 30 s
forno1.desligar()
forno1.aquecer(10)
````

### Métodos Dunder
Vamos tentar executar uma operação de soma de dois fornos de micro-ondas...

````python
forno2 = FornoMicroOndas("Electrolux", "C", 40)
print(forno1 + forno2)
````

Como podemos ver pela mensagem de erro da célula anterior, a operação de soma '+' não é suportada para objetos do tipo `FornoMicroOndas`. O que podemos fazer é implementar o método **Dunder** (_double underscores_) `__add__()` para implementar a funcionalidade de soma para objetos dessa classe.

````python
# Definindo a Classe
class FornoMicroOndas:
    # O método __init__ é o construtor do objeto, e define os atributos que cada objeto terá
    def __init__(self, fabricante, catConsumo, volume):
        self.fabricante = fabricante      # atributo
        self.catConsumo = catConsumo      # atributo
        self.volume = volume              # atributo
        self.ligado = False               # atributo

    # Definindo métodos (comportamento)
    def ligar(self):
        if self.ligado:
            print(f"Forno micro-ondas ({self.fabricante}) já está ligado!")
        else: 
            self.ligado = True
            print(f"Forno micro-ondas ({self.fabricante}) agora está ligado.")

    def desligar(self):
        if self.ligado:
            self.ligado = False
            print(f"Forno micro-ondas ({self.fabricante}) agora está desligado.")
        else: 
            print(f"Forno micro-ondas ({self.fabricante}) já está desligado!")

    def aquecer(self, segundos):
        if self.ligado:
            print(f"Aquecendo ({self.fabricante}) por {segundos} seg.")
        else:
            print(f"Vc precisa ligar o forno micro-ondas ({self.fabricante}) primeiro!")

    def __add__(self, other):
        return f"Volume conjunto dos fornos {self.fabricante} e {other.fabricante}: {self.volume + other.volume} L."

# Criando duas instâncias (objetos) da classe FornoMicroOndas: forno1 e forno2
forno1 = FornoMicroOndas("Philco", "B", 22)
forno2 = FornoMicroOndas("Electrolux", "C", 40)

# somando dois fornos
print(forno1 + forno2)

forno3 = FornoMicroOndas("Panasonic", "A", 32)
print(forno1 + forno2 + forno3)
````

### Exercício Final:
Como resolver o problema da soma do volume de mais de <u>dois fornos</u> (descomente as duas últimas linhas do script da célula anterior?  

A mensagem de erro se refere ao fato de que concatenação de 'str' só pode ser realizada com outra 'str' e não com outro objeto da classe "FornoMicroOndas"... Podemos alterar o método dunder `__add__()` para retornar a soma dos volumes de forma numérica, como mostrado a seguir:

#### Função com Quantidade Indeterminada de Parâmetros
Quando não se sabe exatamente quantos parâmetros serão passados a uma determinada função, utilizamos `*args` (para indicar múltiplos valores) ou `**kwargs` (para múltiplos valores nomeados: chave-valor).

##### Exemplo
Soma de diversos valores passados à função `somar_todos()`.

````python
def somar_todos(*numeros):
    total = 0
    for num in numeros:
        total += num
    return total

print(somar_todos(10, 20, 30))    # Resultado: 60
print(somar_todos(2, 4))          # Resultado: 6
````

### Gabarito-1:
Nesta solução vamos gerar um novo objeto da classe <FornoMicroOndas> no método `__add__()` de modo que a linguagem Python vai encontrar o que ela esperava, ou seja, outro objeto da refirada classe.

Para somar três ou mais instâncias de uma classe usando o método **dunder** `__add__()`, não precisamos escrever um método especial para receber três ou mais argumentos. Python, automaticamente encadeia o operador `+` avaliando sequencialmente a expressão `item1 + item2 + item3` como `(item1.__add__(item2)).__add__(item3)`.

````python
# Gabarito 1:  
class FornoMicroOndas:
    # O método __init__ é o construtor do objeto, e define os atributos que cada objeto terá
    def __init__(self, fabricante="", catConsumo="", volume=0):
        self.fabricante = fabricante      # atributo
        self.catConsumo = catConsumo      # atributo
        self.volume = volume              # atributo
        self.ligado = False               # atributo

    # Definindo métodos (comportamento)
    def ligar(self):
        if self.ligado:
            print(f"Forno micro-ondas ({self.fabricante}) já está ligado!")
        else: 
            self.ligado = True
            print(f"Forno micro-ondas ({self.fabricante}) agora está ligado.")

    def desligar(self):
        if self.ligado:
            self.ligado = False
            print(f"Forno micro-ondas ({self.fabricante}) agora está desligado.")
        else: 
            print(f"Forno micro-ondas ({self.fabricante}) já está desligado!")

    def aquecer(self, segundos):
        if self.ligado:
            print(f"Aquecendo ({self.fabricante}) por {segundos} seg.")
        else:
            print(f"Vc precisa ligar o forno micro-ondas ({self.fabricante}) primeiro!")

    def __add__(self, other):
        return FornoMicroOndas(volume=self.volume + other.volume)
    
    def __str__(self):
        return f"{self.volume}"
  
# Criando quatro instâncias (objetos) da classe FornoMicroOndas
forno1 = FornoMicroOndas("Philco","B",15)
forno2 = FornoMicroOndas("Electrolux","C",40)
forno3 = FornoMicroOndas("Panasonic", "A", 30)
forno4 = FornoMicroOndas("Mondial", "C", 15)
print(f"Capacidade volumétrica conjunta dos 4 fornos: {forno1 + forno2 + forno3 + forno4} L.")
````

Outra possibilidade é que podemos ter somas com constantes: `forno1 + 10` ou `10 + forno1`. Neste caso precisaremos usar o método `__radd__()` para resolver esse questão. Basicamente, no caso da adição falhar como `10 + forno1`), ele tentará a mesma operação com operandos reversos, como `forno1 + 10`.

````python
# Gabarito-2
class FornoMicroOndas:
    # O método __init__ é o construtor do objeto, e define os atributos que cada objeto terá
    def __init__(self, fabricante="", catConsumo="", volume=0):
        self.fabricante = fabricante      # atributo
        self.catConsumo = catConsumo      # atributo
        self.volume = volume              # atributo
        self.ligado = False               # atributo

    # Definindo métodos (comportamento)
    def ligar(self):
        if self.ligado:
            print(f"Forno micro-ondas ({self.fabricante}) já está ligado!")
        else: 
            self.ligado = True
            print(f"Forno micro-ondas ({self.fabricante}) agora está ligado.")

    def desligar(self):
        if self.ligado:
            self.ligado = False
            print(f"Forno micro-ondas ({self.fabricante}) agora está desligado.")
        else: 
            print(f"Forno micro-ondas ({self.fabricante}) já está desligado!")

    def aquecer(self, segundos):
        if self.ligado:
            print(f"Aquecendo ({self.fabricante}) por {segundos} seg.")
        else:
            print(f"Vc precisa ligar o forno micro-ondas ({self.fabricante}) primeiro!")

    def __add__(self, other):
        if isinstance(other, FornoMicroOndas):
            return FornoMicroOndas(volume=self.volume + other.volume)
        elif isinstance(other, int) or isinstance(other, float):
            return FornoMicroOndas(volume=self.volume + other)
        else:
            raise TypeError("Tipo de operando(s) não suportados para '+'.")
        
    def __radd__(self, other):
        return self.__add__(other)
    
# Criando quatro instâncias (objetos) da classe FornoMicroOndas
forno1 = FornoMicroOndas("Philco","B",15)
forno2 = FornoMicroOndas("Electrolux","C",40)
forno3 = FornoMicroOndas("Panasonic", "A", 30)
forno4 = FornoMicroOndas("Mondial", "C", 15)
print(f"Capacidade volumétrica conjunta dos 4 fornos: {(10 + forno1 + forno2 + forno3 + forno4 + 25).volume} L.")
````
