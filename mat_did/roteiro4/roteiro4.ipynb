<u>Definição</u>: A **Realimentação** (*feedback*) no controle de processos é um **mecanismo reativo** que avalia dados de saída para medir o desempenho em relação a padrões pré-determinados, identificar variâncias e **aplicar ações corretivas**. Ela atua como um ciclo contínuo para estabilizar processos, corrigir erros e garantir que os objetivos sejam alcançados.

 ### <span style="color: #0040ff;">1. Necessidade da Realimentação</span>

Em geral, os sistemas são projetados para produzir uma saída $y(t)$ desejada para uma dada entrada $x(t)$. Os sistemas projetados com **malha aberta**, na prática sofrem alterações de suas características com o passar do tempo, alterando os seus parâmetros com o envelhecimento e até com a substituição de alguns componentes, ou mesmo com mudanças no ambiente no qual o sistema é operado. Estas **variações** causam
mudanças na saída ainda que se repita uma determinada entrada. Obviamente, isto não é desejável para sistemas de precisão...

Para mitigar as variações das características podemos acrescentar à entrada um sinal que não dependa do tempo (para realizar correção) e que terá a missão de contrabalançar os efeitos das variações das características do sistema e das variáveis ambientais.

Como essas mudanças no sistema e no ambiente são geralmente imprevisíveis, não é evidente como vamos programar as correções apropriadas na entrada. Porém, a **diferença** entre a **saída atual** e a **saída desejada** nos fornece uma indicação da **correção** adequada a ser aplicada à entrada do sistema. Agindo dessa maneira podemos contrabalançar as variações através da alimentação da saída de volta à entrada (**realimentação**).

### <span style="color: #0040ff;">2. Exemplo de Sistema com Realimentação</span>

Qual é o preço ótimo de um produto? Evidentemente que é aquele que maximiza o lucro...  

Nesse caso, o sistema tem como saída o **lucro** e como entrada o **preço do produto**. A saída (lucro) pode ser controlada (dentro de certos limites) variando-se a entrada (preço):
- Inicialmente o comerciante pode colocar um preço muito alto. Mas ele venderá poucos itens, reduzindo o lucro.
- Se o preço for baixo, as vendas subirão, mas o lucro por unidade não será maximizado pois o preço de venda estará próximo do valor de custo de cada unidade.

O comerciante pode usar a **realimentação** do lucro (saída) para ajustar o preço (entrada) de modo a maximizar seu lucro. 

Por exemplo, se houver uma mudança repentina no mercado, tal como uma greve que fecha uma grande fábrica na cidade, a demanda pelo item reduzirá, diminuindo a saída (lucro). O comerciante ajusta a entrada (reduz o preço) usando a realimentação da saída (lucro) de forma a otimizar o lucro considerando a circunstância de fechamento da fábrica. Por outro lado, se a cidade repentinamente se tornar mais próspera devido a uma nova fábrica inaugurada, nesse caso, o comerciante aumentará o preço para maximizar o lucro. 

Portanto, através da **realimentação** contínua da saída para a entrada, o comerciante consegue atingir seu objetivo de lucro máximo (saída otimizada) em qualquer circunstância...

**Exemplo**: Diagrama de blocos de um sistema realimentado (malha fechada).
<img src="rot4_fig1.png" alt="Sistema Realimentado" width="400" style="display: block; margin: 0 auto;">  

Vamos dizer que o sistema realimentado mostrado na figura anterior seja um **Amplificador de Áudio**, com um ganho de caminho direto (malha aberta), $G = 10.000$; e vamos adotar uma realimentação de um centésimo da saída para a entrada, $H = 0,01$.  

<img src="rot4_fig2.png" alt="Sistema Realimentado - Amplificador de Áudio" width="400" style="display: block; margin: 0 auto;">  

Pode-se demonstrar que o ganho total, $T(s)$, do amplificador realimentado é dado por:
$$T(s)=\frac{G(s)}{1+G(s).H(s)} = \frac{10^4}{1+10^4\times 10^{-2}} = \frac{10^4}{101} \approx 99,01$$

Agora vamos supor que, devido a um longo tempo de uso ou a uma substituição de um transistor do amplificador, o ganho direto $G(s)$ dobre, mudando de 10.000 para 20.000. O novo ganho total do amplificador realimentado será:
$$T(s)=\frac{G(s)}{1+G(s).H(s)} = \frac{2 \times 10^4}{1+200} \approx 99,5$$

Surpreendentemente, uma variação de 100% no ganho direto do amplificador com realimentação causa uma variação de apenas 0,5% no ganho total $T(s)$. Esta **redução na sensibilidade a variações paramétricas** é a principal característica desejável nos amplificadores de precisão.  

#### <span style="color: #ab2c02;">Conclusão</span>
<span style="color: #ab2c02;">No exemplo do **Amplificador de Áudio Realimentado**, reduzimos a sensibilidade do ganho do amplificador a variações paramétricas ao custo do ganho da malha aberta (ganho direto), reduzido de **10.000** para **99** no sistema de malha fechada. A baixa sensibilidade à variação de um parâmetro, nesse caso do **ganho direto**, é extremamente importante em sistemas de precisão.

### <span style="color: #0040ff;">3. Realimentação Negativa e Realimentação Positiva</span>

No exemplo anterior, amplificador de áudio, temos um sistema com **realimentação negativa**, ou seja, o sinal da saída é multiplicado pelo ganho da realimentação, $H(s)=0,01$, e chega ao bloco de soma com sinal negativo, sendo subtraído da entrada $x(t)$.

Se o sinal da saída passar pelo ganho da realimentação e ser somado com a entrada $x(t)$, então dizemos que o sistema tem **realimentação positiva**, resultando num ganho total dado por:

$$T(s)=\frac{G(s)}{1-G(s).H(s)} $$

Novamente, vamos supor $G(s)=10.000$ e $H(s)=0,9 \times 10^{-4}$, então o ganho total será:
$$T(s)=\frac{10^4}{1 - 10^4 . 0,9\times 10^{-4}} = 10^5 = 100.000$$

Agora vamos supor que devido ao tempo de uso ou a substituição de alguns transistores, o ganho de malha direta do amplificador mude para $11.000$. O novo ganho do amplificador realimentado será:
$$T(s)=\frac{11.000}{1 - 11.000 . 0,9\times 10^{-4}} = 1.100.000$$

Observe que, neste caso, um simples aumento de 10% no ganho direto $G(s)$ resultou em um aumento de 1.100% no ganho do sistema com realimentação positiva $T(s)$ (de 100.000 para 1.100.000). Claramente, o amplificador tornou-se muito sensível a variações paramétricas. Este comportamento é exatamente oposto ao que foi observado anteriormente, quando o sinal realimentado é subtraído da entrada $x(t)$.

#### <span style="color: #ab2c02;">Conclusão</span>
<span style="color: #ab2c02;">A **realimentação positiva** aumenta o ganho do sistema, mas tende a tornar o sistema mais sensível a variações paramétricas.  
Ela também pode resultar em instabilidade do sistema.</span>

No exemplo do Amplificador de Áudio com **realimentação positiva**, se $G(s) = 11.111$, então $G(s).H(s) = 1$ e $T(s) \to \infty$ e o sistema se torna instável, pois o sinal de realimentação é exatamente igual ao próprio sinal de entrada, pois $G(s).H(s) = 1$. Quando o sinal de entrada é aplicado, não importa quão pequeno ou curto em duração ele seja, ele será realimentado inteiramente e reforçará a entrada, a qual passará para a saída novamente, sendo realimentada de novo, e de novo, e de novo..., o sinal se perpetuará indefinidamente. Essa perpetuação, mesmo que a entrada deixe de existir, é precisamente um sintoma de **instabilidade**.

### <span style="color: #0040ff;">4. Análise de um Sistema de Controle Simples</span>

Modelo matemático de um motor DC:
<img src="rot4_fig5.png" alt="Modelo matemático de um motor DC" width="600" style="display: block; margin: 0 auto;">  

A figura seguinte representa um **Sistema de Controle de Posição Automático** (realimentado), utilizado geralmente no controle da posição angular de um objeto pesado, tal como: uma antena de radar, uma bateria anti-aérea ou um satélite em órbita. 

<img src="rot4_fig3.png" alt="Sistema de Controle de Posição Automático" width="600" style="display: block; margin: 0 auto;">  

A posição do potenciômetro da entrada, $\theta_i$, é a posição angular desejada do objeto remoto (entrada). A posição angular atual $\theta_o$ do objeto remoto (saída) é medida através de um potenciômetro cujo eixo é comum ao eixo do objeto controlado. A diferença entre a entrada $\theta_i$ (posição desejada do objeto controlado) e a saída $\theta_o$ (posição atual do objeto controlado) é amplificada.  

A saída amplificada que é proporcional à diferença $\theta_i - \theta_o$, é aplicada na entrada do motor DC $(E_m(s))$. Se essa diferença for nula, $\theta_i - \theta_o = 0$ (a saída é igual a entrada, ângulo desejado), não existirá entrada aplicada ao motor e ele para. Mas se $\theta_i \ne \theta_o$, existirá uma entrada não nula no motor, o qual girará seu eixo até que $\theta_i = \theta_o$. É evidente que ajustando o potenciômetro de entrada para uma desejada posição neste sistema, podemos controlar a posição angular de um objeto pesado e remoto.  

A seguir temos o diagrama de blocos do referido sistema, supondo o ganho ajustável do amplificador da diferença como $K$:

<img src="rot4_fig4.png" alt="Sistema de Controle de Posição Automático" width="600" style="display: block; margin: 0 auto;">  

Considere que a **função de transferência do motor DC** (com a carga) que relaciona o ângulo de saída $\theta_o$ com a tensão de entrada do motor DC (servomotor) é $G(s)$. Temos um arranjo de realimentação negativa (ver figura seguinte) com $H(s) = 1$. Logo, $T(s)$, a **função de transferência do sistema realimentado** (malha fechada) que relaciona a saída $\theta_o$ com a entrada $\theta_i$ é:

$$\frac{\Theta_o(s)}{\Theta_i(s)} = T(s) = \frac{K.G(s)}{1+K.G(s)} $$

#### <span style="color: #0040ff;">4.1 Entrada em Degrau Unitário</span>

Se desejarmos mudar a posição angular de um objeto instantaneamente, podemos aplicar uma entrada degrau unitário. A resposta do sistema a essa entrada nos permite esclarecer algumas questões:
1. Quanto tempo o sistema gastará para se posicionar em um novo ângulo ajustado?
1. O sistema realmente atingirá o ângulo ajustado? 
1. O sistema atingirá o ângulo ajustado suavemente (monotonicamente) ou oscilando ao redor da posição final? 
1. Se o sistema oscilar, oscilará por quanto tempo?

Todas estas questões podem ser respondidas determinando-se a saída $\theta_o(t)$ para uma entrada $\theta_i(t) = u(t)$.  
A entrada **degrau unitário** implica em uma mudança instantânea no ângulo do potenciômetro. Esta entrada é uma das mais difíceis de ser seguida por um sistema..., e se o sistema se comportar bem para esta entrada, então ele provavelmente terá um bom comportamento nas outras situações esperadas. Este é o motivo pelo qual **testamos sistemas de controle com entrada degrau unitário**.

Para a entrada em degrau, $\theta_i(t) = u(t)$, temos $\Theta_i(s) = 1/s$ e 
$$\Theta_o(s) = \frac{1}{s} T(s) = \frac{K.G(s)}{s(1+K.G(s))}$$

Considerando a função de transferência do motor (com a carga) que relaciona o ângulo da carga $\theta_o(t)$ com a tensão de entrada (tensão de armadura) do motor, como sendo $G(s) = 1/(s^2 + 8s)$, temos:
$$\Theta_o(s) = \frac{K/(s^2 + 8s))}{s(1+K/(s^2 + 8s))} = \frac{K}{s(s^2+8s+K)}$$

Vamos investigar o comportamento do sistema para três valores diferentes do ganho do amplificador da diferença, $K$:

* Para K = 7: $$\Theta_o(s) = \frac{7}{s(s^2+8s+7)} = \frac{1}{s} - \frac{7/6}{s+1} + \frac{1/6}{s+7}$$
Aplicando Transformada de Laplace Inversa: $$\theta_o(t) = (1-\frac{7}{6}e^{-t}+\frac{1}{6}e^{-7t}).u(t)$$

Esta resposta mostra que o sistema atinge o ângulo desejado (ver figura seguinte), mas a uma taxa bem lenta...

```python
from numpy import linspace, exp, sqrt, cos, pi
import matplotlib.pyplot as plt

def u(t):
    ''' Sinal degrau unitário.
        t : vetor com a base temporal (abscissas)
        Retorna o sinal degrau conforme a base temporal desejada
    '''
    return t >= 0

t = linspace(0, 5, 1000)
theta_o = (1 - 7/6*exp(-t) + exp(-7*t)/6)*u(t)    # saída do sistema realimentado com ganho direto K = 7

plt.figure(figsize=(12,2.5))
plt.plot(t, theta_o)
plt.title("Resposta ao Degrau Unitário para Ganho Direto $K = 7$")
plt.xlabel("tempo (s)"); plt.ylabel(r"$\theta_o(t)$ (rad)")
plt.grid(True)
```

Para acelerar a resposta, vamos aumentar o ganho direto ($K$) para um valor bem maior.

* Para K = 80: $$\Theta_o(s) = \frac{80}{s(s^2+8s+80)} = \frac{1}{s} + \frac{\frac{\sqrt{5}}{4}e^{j153°}}{s+4-j8} + \frac{\frac{\sqrt{5}}{4}e^{-j153°}}{s+4+j8}$$
$$\theta_o(t) = \left [1 + \frac{\sqrt{5}}{2}e^{-4t}.cos(8t+153°) \right ].u(t)$$

Esta resposta atinge a posição final a uma taxa mais rápida do que no caso anterior ($K = 7$) como podemos ver na figura seguinte. Infelizmente, a melhora é obtida ao custo de oscilações com um alto sobressinal (*overshoot*). Nesse caso o sobressinal percentual (ou ultrapassagem percentual -- medida do quanto a saída de um sistema dinâmico ultrapassa seu valor final ou de regime permanente) é de 21%. 

A resposta atinge o valor de pico no **tempo de pico**, $t_p = 0,393$ segundos.  

O **tempo de subida** (tempo necessário para a resposta subir de 10% para 90% do seu valor de regime permanente) indica a velocidade da resposta. No presente caso, $t_r = 0,175$ segundos.  

O valor de **regime permanente da resposta** nesse caso é unitário, tal que o **erro de regime permanente** é zero. Teoricamente, é necessário um tempo infinito para a resposta atingir o valor desejado (unitário). Na prática, entretanto, podemos considerar que a resposta atingiu o valor final se ela estiver muito próxima do valor final. Uma medida amplamente aceita de proximidade é estar dentro de 2% de seu valor final. O tempo necessário para a resposta atingir e permanecer dentro de 2% do valor final é chamado de **tempo de acomodação**, $t_s$ (percentuais utilizados: 2% e 5%). Na figura seguinte vemos $t_s \approx 1$ segundo para $K = 80$.  

<p style="text-align: center;"><b>Um bom sistema apresenta um pequeno sobressinal, um pequeno t<sub>r</sub> e t<sub>s</sub> e um pequeno erro de regime permanente.</b></p>

```python
from numpy import linspace, exp, sqrt, cos, pi
import matplotlib.pyplot as plt

def u(t):
    return t >= 0

t = linspace(0, 5, 1000)
plt.figure(figsize=(12,3))
for K in (7, 16, 80):
    match K:
        case 7: 
            theta_o = (1 - 7/6*exp(-t) + exp(-7*t)/6)*u(t)
        case 80:
            theta_o = (1 + sqrt(5)/2*exp(-4*t)*cos(8*t+153*pi/180))*u(t)
        case 16:
            theta_o = (1 - (4*t+1)*exp(-4*t))*u(t)
    plt.plot(t, theta_o, label=f"K = {K}")
plt.title("Respostas ao Degrau Unitário para vários Ganhos Diretos (K)")
plt.xlabel("tempo (s)"); plt.ylabel(r"$\theta_o(t)$ (rad)"); plt.grid(True)
plt.legend();
```

Um grande sobressinal pode ser inaceitável em várias aplicações.  

Vamos tentar determinar K (o ganho direto) que resulta em uma resposta rápida e sem oscilações.  

Raízes características complexas levam a oscilações. Ou seja, para evitar oscilações, as raízes características devem ser reais. Para o sistema em análise (posicionamento via motor DC e usando potenciômetro como atuador de indicação do ângulo -- tensão de armadura), o polinômio
característico é $s^2 + 8s + K$. Para $K > 16$, as raízes características são complexas. Para $K < 16$, as raízes são reais. A resposta mais rápida e sem oscilações é obtida escolhendo $K = 16$. 

* Para K = 16: $$\Theta_o(s) = \frac{16}{s(s^2+8s+16)} = \frac{1}{s} - \frac{1}{s+4} - \frac{4}{(s+4)^2}$$
$$\theta_o(t) = [1 - (4t+1)e^{-4t}].u(t)$$

O sistema com $K > 16$ é dito ser **subamortecido** (resposta oscilatória), enquanto que o sistema com $K < 16$ é dito ser **superamortecido** (resposta lenta). Para $K = 16$, o sistema é dito ser **criticamente amortecidoo** (resposta mais rápida e sem oscilações).

Existe um compromisso entre um sobressinal indesejado e o tempo de subida. Reduzindo o sobressinal aumenta-se o tempo de subida (sistema com resposta mais lenta). Na prática, um pequeno sobressinal pode ser aceitável, por ser mais rápido do que o amortecimento crítico. Note que o sobressinal percentual (SSP) e o tempo de subida $t_r$ não possuem significado para os casos de superamortecido e amortecimento crítico. Além de ajustar o ganho $K$, podemos precisar aumentar o sistema com algum tipo de compensador se as especificações de sobressinal e velocidade de resposta forem muito restritivas.

#### <span style="color: #0040ff;">4.2 Entrada em Rampa</span>

Suponha que o sistema de controle do exemplo das seções anteriores guie uma bateria anti-aérea. Suponha também que se está rastreando um avião inimigo que está se movendo com velocidade uniforme, ou seja, o ângulo de posicionamento da bateria deve aumentar linearmente com o tempo $t$. Logo, a entrada, neste caso, será uma rampa, $\theta_i(t) = t.u(t)$. 

Vamos definir a resposta do sistema a esta entrada quando $K = 80$. Neste caso, $\Theta_i(s) = 1/s^2$ e
$$\Theta_o(s) = \frac{80}{s^2(s^2+8s+80)} = -\frac{0.1}{s} + \frac{1}{s^2} + \frac{0.1(s-2)}{s^2+8s+80}$$
$$\theta_o(t) = [-0.1 + t + \frac{1}{8}e^{-8t}.cos(8t+36,87°)].u(t)$$

Esta resposta, mostrada na figura seguinte, mostra que existe um **erro de regime permanente** $e_rp = 0,1$ radianos. Em vários casos, este pequeno erro de regime permanente é tolerável. Se, entretanto, um erro de regime permanente nulo para entrada em rampa for necessário, este sistema em sua forma atual é insatisfatório. Devemos adicionar alguma forma de compensação ao sistema.

```python
t = linspace(0, 0.8, 200)
theta_o = (-0.1 + t + 1/8.*exp(-8*t)*cos(8*t + 36.87*pi/180))*u(t)
plt.figure(figsize=(6, 3))
plt.plot(t, theta_o, label="Real")
plt.plot(t, t, 'b:', label="Desejado")
plt.title("Resposta à rampa com K = 80")
plt.xlabel("tempo (s)"); plt.ylabel(r"$\theta_o(t)$ (rad)"); 
plt.legend(); plt.grid(True);
```
