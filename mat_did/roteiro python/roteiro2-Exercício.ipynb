### Exercício Final

Criar em Python o <b>Jogo de Advinhação de um Número Inteiro, entre 1 e 100</b>, com as seguintes entradas e saídas: 
 
    Estou pensando em um número!  
    Tente adivinhar o número em que estou pensando: 25
   
    Muito baixo! Adivinhe novamente: 50  
    Muito alto! Adivinhe novamente:  42
   
    É isso! Você acertou!  
    Quer jogar novamente? (sim/não) não  
    Obrigado por jogar! 

```python
import random

print("tente adivinhar um número de 0 até 100")
result = 0
randomnum = random.randint(0, 100)
while result == 0:
    num = int(input("adicione um número"))
    if num > 0 and num < 100:
        if num == randomnum:
            print("você acertou!")
            result = 1
        elif num > randomnum:
            print("número muito alto! tente novamente.")
        else:
            print("número muito baixo! tente novamente.")
    else:
        print("número fora da faixa, adicione um número entre 0 e 100!")
```
